---

## What is a `.env` File?

A `.env` file stores **configuration variables** (API keys, passwords, database URLs) outside your code.

### Why This Matters

❌ **Bad Practice**:
```python
api_key = "sk-proj-abc123xyz"  # NEVER do this!
```

✅ **Good Practice**:
```python
# In .env file
OPENAI_API_KEY=sk-proj-abc123xyz

# In code
import os
from dotenv import load_dotenv
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
```

### Why?

1. **Security**: Keys not exposed in git history or source code
2. **Portability**: Different keys for dev/staging/production without changing code
3. **Industry Standard**: Every company uses `.env` or similar (AWS Secrets Manager, Vault, etc.)

````python
# Demo: Loading .env file

import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Retrieve API key from environment
openai_api_key = os.getenv("OPENAI_API_KEY")

# Check if key exists (never print full key in production!)
if openai_api_key:
    # Show masked version for demonstration
    masked_key = openai_api_key[:10] + "..." + openai_api_key[-5:]
    print(f"✅ API Key loaded successfully: {masked_key}")
else:
    print("❌ API Key not found. Make sure .env file exists and OPENAI_API_KEY is set.")
````

## Part 2.3: Streaming Responses

Streaming lets you show responses **as they're generated** (like ChatGPT typing effect).

````python
# Demo: Streaming responses

print("🔄 Streaming response (typed effect):\n")

stream = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": "Explain neural networks in 3 sentences."}],
    stream=True,  # Enable streaming
)

full_response = ""
for chunk in stream:
    # Each chunk contains a token delta
    if chunk.choices[0].delta.content is not None:
        token = chunk.choices[0].delta.content
        print(token, end="", flush=True)  # Print without newline
        full_response += token

print("\n")
print(f"\n✅ Full response received ({len(full_response)} characters)")
````

## Part 2.4: Function Calling (Tools)

Function calling lets you tell the LLM: "Here are tools you can use. Use them when needed."

### Use Cases
- **Structured data**: Extract info as JSON
- **Tool use**: Access APIs (weather, database, calculator)
- **Agent behavior**: LLM decides what to do step-by-step

````python
# Demo: Define functions (tools) the model can use

tools = [
    {
        "type": "function",
        "function": {
            "name": "get_current_weather",
            "description": "Get the current weather in a given location",
            "parameters": {
                "type": "object",
                "properties": {
                    "location": {
                        "type": "string",
                        "description": "The city and state, e.g. San Francisco, CA"
                    },
                    "unit": {
                        "type": "string",
                        "enum": ["celsius", "fahrenheit"],
                        "description": "Temperature unit"
                    }
                },
                "required": ["location"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "calculate_sum",
            "description": "Add two numbers together",
            "parameters": {
                "type": "object",
                "properties": {
                    "number_1": {"type": "number", "description": "First number"},
                    "number_2": {"type": "number", "description": "Second number"}
                },
                "required": ["number_1", "number_2"]
            }
        }
    }
]

print("✅ Tools defined. Now let's ask the model to use them.")
````

````python
# Demo: Model chooses to use a function

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": "What is 42 plus 58?"}],
    tools=tools,
    tool_choice="auto",  # Let model decide when to use tools
)

print("📋 Response from model:")
choice = response.choices[0]

if choice.message.tool_calls:
    for tool_call in choice.message.tool_calls:
        print(f"\n🔧 Model wants to call: {tool_call.function.name}")
        print(f"   Arguments: {tool_call.function.arguments}")
        
        # Parse arguments
        args = json.loads(tool_call.function.arguments)
        
        # Execute the function
        if tool_call.function.name == "calculate_sum":
            result = args["number_1"] + args["number_2"]
            print(f"   ✅ Result: {result}")
else:
    print(f"📝 Model response: {choice.message.content}")
````

# Topic 3: Tokenization Deep Dive 🔤

**Objective**: Understand how LLMs break down text into tokens, count them, and estimate costs.

---

## What is a Token?

A **token** is a small chunk of text. LLMs process text **one token at a time**.

### Examples

- "Hello" = 1 token
- "Hello, world!" = 4 tokens ("Hello", ",", "world", "!")
- "🚀" = 1 token
- "ing" = 1 token (part of "running" = "run" + "ning" = 2 tokens)

### Why Tokenization?

1. **Cost model**: You pay per 1000 tokens, not per word
2. **Context limits**: GPT-3.5 has 4K context; GPT-4 has 128K (all measured in tokens)
3. **Model efficiency**: Subword tokenization handles rare words better than word-level

### Token Budget Example

```
Prompt: 500 tokens
Response: 300 tokens
Total: 800 tokens ≈ $0.001-0.003 (depending on model)
```

````python
# Install tiktoken (OpenAI's tokenizer)
# !pip install tiktoken

import tiktoken

# Load tokenizer for GPT-3.5-turbo
encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")

print("✅ Tokenizer loaded")
````

````python
# Demo: Basic tokenization

sample_text = "Hello, world! This is an AI."

# Tokenize
tokens = encoding.encode(sample_text)

print(f"📝 Text: {sample_text}")
print(f"🔢 Tokens: {tokens}")
print(f"📊 Token count: {len(tokens)}")
print()

# Decode back
decoded = encoding.decode(tokens)
print(f"✅ Decoded: {decoded}")
````

````python
# Demo: Visual breakdown of tokenization

sample_sentences = [
    "The cat sat on the mat.",
    "Machine learning is amazing!",
    "🚀 Let's tokenize emoji!",
]

print("📊 Tokenization Breakdown\n")
print("-" * 70)

for sentence in sample_sentences:
    tokens = encoding.encode(sentence)
    print(f"\n📝 Sentence: {sentence}")
    print(f"🔢 Tokens ({len(tokens)}): {tokens}")
    
    # Show each token decoded
    token_strings = [encoding.decode([t]) for t in tokens]
    print(f"🔤 Decoded: {token_strings}")
````

# Topic 4: Embeddings & Semantic Search 🧮

**Objective**: Learn what embeddings are, how to create them, and build a simple semantic search.

---

## What Are Embeddings?

An **embedding** is a list of numbers (vector) that represents the **meaning** of text.

### The Core Idea

```
Text: "The cat sat on the mat."
    ↓
Embedding: [-0.012, 0.087, 0.052, ..., 0.091]  (1536 numbers for GPT-3 embedding model)
```

**Key property**: Similar meanings → similar embeddings (vectors close in space)

```
"The cat sat on the mat." → [0.1, 0.2, 0.3, ...]
"A feline sat on a rug."   → [0.11, 0.19, 0.31, ...]
                     (similar, close distance)

"The cat sat on the mat." → [0.1, 0.2, 0.3, ...]
"Quantum computing rocks." → [0.8, -0.1, 0.9, ...]
                     (dissimilar, far distance)
```

### Use Cases

1. **Semantic Search**: "Find documents similar to my query"
2. **Clustering**: "Group similar texts together"
3. **Recommendation**: "Suggest items similar to this one"
4. **Anomaly Detection**: "This embedding is far from all others"

````python
# Demo: Create embeddings

texts = [
    "The cat is sleeping on the couch.",
    "A dog is playing in the park.",
    "Machine learning models need data.",
    "Neural networks are inspired by the brain.",
]

# Create embeddings using OpenAI's API
embeddings = []

for text in texts:
    response = client.embeddings.create(
        model="text-embedding-3-small",  # Fast and cheap
        input=text
    )
    embedding = response.data[0].embedding
    embeddings.append(embedding)
    print(f"✅ Embedded: {text[:40]}...")

print(f"\n📊 Embedding dimension: {len(embeddings[0])}")
print(f"   First 5 values: {embeddings[0][:5]}")
````

````python
# Demo: Compute similarity between embeddings

import numpy as np

def cosine_similarity(a, b):
    """Compute cosine similarity between two vectors"""
    dot_product = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot_product / (norm_a * norm_b)

# Compare first and second text
similarity_1_2 = cosine_similarity(embeddings[0], embeddings[1])
similarity_1_3 = cosine_similarity(embeddings[0], embeddings[2])

print("🔍 Similarity Scores\n")
print(f"Text 1: {texts[0]}")
print(f"Text 2: {texts[1]}")
print(f"  Similarity: {similarity_1_2:.4f}\n")

print(f"Text 1: {texts[0]}")
print(f"Text 3: {texts[2]}")
print(f"  Similarity: {similarity_1_3:.4f}\n")

print("📊 Interpretation:")
print(f"  Animals are similar: {similarity_1_2:.4f}")
print(f"  Animal vs AI: {similarity_1_3:.4f}")
print("  → Lower similarity for different topics!")
````

````python
# Demo: Semantic search - Find most similar text

def semantic_search(query, documents, embeddings):
    """Find the most similar document to a query"""
    # Embed the query
    query_response = client.embeddings.create(
        model="text-embedding-3-small",
        input=query
    )
    query_embedding = query_response.data[0].embedding
    
    # Compute similarity with all documents
    similarities = [cosine_similarity(query_embedding, doc_emb) for doc_emb in embeddings]
    
    # Find most similar
    most_similar_idx = np.argmax(similarities)
    
    return most_similar_idx, similarities

# Test semantic search
query = "A kitten playing with yarn"
idx, scores = semantic_search(query, texts, embeddings)

print(f"🔍 Query: {query}\n")
print("📊 Similarity scores:")
for i, (text, score) in enumerate(zip(texts, scores)):
    marker = "⭐" if i == idx else "  "
    print(f"{marker} [{score:.4f}] {text}")
    
print(f"\n✅ Most similar: Text {idx + 1}")
````

# Topic 5: Attention Mechanism Explained 👀

**Objective**: Understand how transformers focus on relevant parts of text using attention.

---

## What is Attention?

**Intuition**: When you read a sentence, you focus on relevant words.

Example: "The cat sat on the mat."
- Reading "sat": Focus on "cat" (subject) and "mat" (location)
- Reading "mat": Focus on "sat" (verb) and "the" (article)

**Attention mechanism does this automatically**: It learns what to focus on!

### The Math (Simplified)

```
For each word, compute:
1. Query (Q): "What am I looking for?"
2. Key (K): "What am I?"
3. Value (V): "What information do I have?"

Attention Score = softmax(Q · K^T / √d_k) · V
                 ↑           ↑        ↑      ↑
              normalize  query-key  scaling  get info
```

Don't worry about the formula—the key idea is: **align queries with keys, then extract values**.

````python
# Demo: Visualizing attention weights

print("\n" + "="*60)
print("ATTENTION WEIGHTS VISUALIZATION")
print("="*60)

sentence = "The cat sat on the mat"
words = sentence.split()

print(f"\n📝 Sentence: {sentence}")
print(f"🔤 Words: {words}\n")

# Simulate attention weights for the word "sat"
# (In reality, these come from the neural network)
word_focus = "sat"
word_idx = words.index(word_focus)

# Simulated attention: focus on nearby words (especially subject "cat" and location "mat")
attention_weights = {
    "The": 0.05,
    "cat": 0.35,    # Subject - high attention
    "sat": 0.10,    # Self
    "on": 0.15,
    "the": 0.10,
    "mat": 0.25,    # Location - high attention
}

print(f"When processing '{word_focus}':")
print("\nAttention Weights (where the model focuses):\n")
for word, weight in attention_weights.items():
    bar = "█" * int(weight * 40)  # Visual bar
    print(f"{word:6s} {weight:.2f} │{bar}")

print("\n📊 Interpretation:")
print("   - 'cat' has high weight: grammatical subject")
print("   - 'mat' has high weight: location context")
print("   - 'on' has moderate weight: preposition")
print("   - Total weights sum to 1.0 (normalized)")
````

````python
# Demo: Query-Key-Value mechanism

print("\n" + "="*70)
print("QUERY-KEY-VALUE MECHANISM BREAKDOWN")
print("="*70)

print("""
Step 1: Create Q, K, V for each word
────────────────────────────────────

For "The cat sat on the mat":

Word     Query (Q)              Key (K)                Value (V)
────     ─────────              ─────────              ─────────
The      [0.2, 0.1]             [0.1, 0.2]             Info about "The"
cat      [0.5, 0.3]             [0.4, 0.5]             Info about "cat"
sat      [0.3, 0.6]             [0.6, 0.3]             Info about "sat"
on       [0.1, 0.4]             [0.2, 0.1]             Info about "on"
the      [0.2, 0.1]             [0.1, 0.2]             Info about "the"
mat      [0.4, 0.2]             [0.3, 0.4]             Info about "mat"

Step 2: Match queries to keys
──────────────────────────────

Q(sat) · K(all words)^T = [scores]

sat Q · cat K     = 0.3*0.4 + 0.6*0.5 = 0.42 ← HIGH (relevant!)
sat Q · mat K     = 0.3*0.3 + 0.6*0.4 = 0.33 ← HIGH (relevant!)
sat Q · The K     = 0.3*0.1 + 0.6*0.2 = 0.15 ← LOW (less relevant)

Step 3: Softmax normalization (convert to probabilities)
─────────────────────────────────────────────────────────

Higher scores → higher probability

Step 4: Extract information from values
───────────────────────────────────────

Output = attention_weights · Value(all words)
       = 0.35 * V(cat) + 0.25 * V(mat) + 0.15 * V(on) + ...
       = Weighted combination of all values
""")
````

````python
# Demo: Implement scaled dot-product attention in NumPy

def scaled_dot_product_attention(Q, K, V):
    """
    Compute scaled dot-product attention.
    
    Args:
        Q: Query matrix (shape: seq_len x d_k)
        K: Key matrix (shape: seq_len x d_k)
        V: Value matrix (shape: seq_len x d_v)
    
    Returns:
        Output: Attention output (shape: seq_len x d_v)
        Attention_weights: Normalized attention scores
    """
    import numpy as np
    
    d_k = K.shape[-1]
    
    # Step 1: Compute Q · K^T / sqrt(d_k)
    scores = np.matmul(Q, K.T) / np.sqrt(d_k)
    
    # Step 2: Apply softmax to get probabilities
    def softmax(x):
        e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
        return e_x / np.sum(e_x, axis=-1, keepdims=True)
    
    attention_weights = softmax(scores)
    
    # Step 3: Apply attention to values
    output = np.matmul(attention_weights, V)
    
    return output, attention_weights

# Example: Simple 4-word sequence, 2D embeddings
print("\n" + "="*60)
print("SCALED DOT-PRODUCT ATTENTION IMPLEMENTATION")
print("="*60 + "\n")

# Create small random embeddings
np.random.seed(42)  # For reproducibility

seq_len = 4  # Number of words
d_k = 3      # Embedding dimension

Q = np.array([
    [0.5, -0.2, 0.1],  # Query for word 1
    [0.3,  0.4, -0.1], # Query for word 2
    [-0.1, 0.6, 0.2],  # Query for word 3
    [0.2, -0.3, 0.5],  # Query for word 4
])

K = np.array([
    [0.1, 0.2, -0.3],  # Key for word 1
    [0.4, -0.1, 0.2],  # Key for word 2
    [-0.2, 0.5, 0.1],  # Key for word 3
    [0.3, 0.1, -0.4],  # Key for word 4
])

V = np.array([
    [1.0, 0.0],  # Value for word 1
    [0.0, 1.0],  # Value for word 2
    [0.5, 0.5],  # Value for word 3
    [-0.5, 0.5], # Value for word 4
])

# Compute attention
output, weights = scaled_dot_product_attention(Q, K, V)

print("Input Shapes:")
print(f"  Q (Queries): {Q.shape}")
print(f"  K (Keys): {K.shape}")
print(f"  V (Values): {V.shape}\n")

print("Attention Weights (normalized, row-wise):")
print(weights)
print(f"\nRow sums (should be 1.0): {weights.sum(axis=1)}\n")

print("Output (weighted combination of values):")
print(output)

print("\n📊 Interpretation:")
print("  - Each row is the context-weighted representation")
print("  - High attention weights → strong influence from that word")
print("  - Output captures both local and distant context")
````
