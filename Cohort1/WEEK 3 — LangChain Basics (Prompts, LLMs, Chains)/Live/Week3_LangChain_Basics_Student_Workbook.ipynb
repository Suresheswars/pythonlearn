## 1) What is LangChain?

LangChain is a framework for building LLM apps step by step.

For beginners, think of it as a toolbox that helps you:
- organize prompts
- call models in a consistent way
- connect multiple steps together
- reuse the same logic in different projects

## 1b) LLM vs Chat Model

**LLM (Language Model):**
- Takes a text prompt, returns text completion
- No concept of "roles" or message structure

**Chat Model:**
- Takes messages with roles (system, user, assistant)
- Returns a message response
- Designed for conversation
- Better for structured workflows

**In this class:** we use chat models because they are clearer and match real-world LLM apps better.

## 2) Must-Know Glossary

- **Prompt:** the instruction you send to the model
- **PromptTemplate:** a reusable prompt with placeholders
- **LLM:** a model that generates text
- **Chat model:** a model that works with messages
- **System message:** the hidden instruction or role
- **User message:** the user's input
- **Chain:** one step feeding into the next step
- **Token:** a chunk of text the model reads or writes
- **Context window:** the maximum amount of text the model can handle at once

## Chains & Orchestration: Translate → Summarize

````python
# Step 1: Define reusable prompt templates
translate_template = 'Translate the following text from {source_lang} to {target_lang}:\n\n{text}\n\nProvide only the translated text.'
summarize_template = 'Summarize the following text in one short paragraph:\n\n{text}\n'

# Step 2: Create a generic LLM calling function with error handling
def call_llm(prompt_text):
    try:
        resp = chat.invoke(prompt_text)
        return getattr(resp, 'content', str(resp))
    except Exception as e:
        return f'LLM error: {e}'

# Step 3: Create specific functions that wrap the templates
def translate(text, source_lang='English', target_lang='Hindi'):
    p = translate_template.format(text=text, source_lang=source_lang, target_lang=target_lang)
    return call_llm(p)

def summarize(text):
    p = summarize_template.format(text=text)
    return call_llm(p)

# Step 4: Test the functions
sample = 'LangChain makes it easier to build reliable LLM-based apps.'

# TODO: Uncomment to test
translated = translate(sample, source_lang='English', target_lang='Hindi')
print('--- Translated ---')
print(translated)

summary = summarize(translated)
print('\n--- Summary of translation ---')
print(summary)
````

### Exercise 4: Structured output (5 min)

In a new cell:
- Write a prompt that asks for JSON output (e.g., name, age, email)
- Parse the response using `json.loads()`
- Handle the case where parsing fails (use try-except)
- Goal: extract structured data from LLM responses

````python
# Exercise 4: Structured output (JSON parsing)
import json
import re

# Step 1: Create a prompt that asks for JSON output
prompt = """Extract the following information from this text and return as JSON:
Text: "My name is Alice, I'm 28 years old, and my email is alice@example.com"

Return a JSON object with keys: name, age, email
Respond ONLY with the JSON object, no other text."""

# Step 2: Call chat.invoke() with the prompt
print("=== JSON Parsing Exercise ===\n")
resp = chat.invoke(prompt)
response_text = resp.content

print(f"Raw response from LLM:\n{response_text}\n")

# Step 3: Extract the response content
# Step 4: Try to parse it as JSON
try:
    # Remove markdown code block markers if present
    cleaned_text = response_text.strip()
    if cleaned_text.startswith('```'):
        # Remove ```json or ``` from start
        cleaned_text = re.sub(r'^```(?:json)?\n?', '', cleaned_text)
        # Remove ``` from end
        cleaned_text = re.sub(r'\n?```$', '', cleaned_text)
    
    parsed_data = json.loads(cleaned_text)
    print("✅ Successfully parsed JSON!")
    print(f"Name: {parsed_data['name']}")
    print(f"Age: {parsed_data['age']}")
    print(f"Email: {parsed_data['email']}")
except json.JSONDecodeError as e:
    print(f"❌ Failed to parse JSON: {e}")
    print("This can happen if the model doesn't return valid JSON.")
    print("Tip: Be more explicit in your prompt about the exact format needed.")
except KeyError as e:
    print(f"❌ Missing key in JSON: {e}")
    print(f"Parsed data: {parsed_data}")
````

## Extension Task: Build a Multi-Step Content Processor

**Goal:** Create a workflow that translates → summarizes → extracts key points (all in one flow).

**Deliverable:** A Python function or notebook that:

1. **Takes a user input:** A block of text in any language
2. **Translates** it to English (if not already)
3. **Summarizes** it in one paragraph
4. **Extracts key points** as a numbered list (structured output)

**Requirements:**
- Use the chat model with system + user messages
- Use prompt templates for all three steps
- Parse the key points as JSON (list of strings)
- Include error handling (API failures, JSON parse errors)
- Test with at least 2 different texts

**Bonus Challenges:**
- Add a `tone` parameter to the summarize step (formal, casual, technical)
- Add a `max_points` parameter to limit key points to N items
- Cache the model responses so you don't re-call for the same input
- Use `batch()` to process multiple texts at once

````python
# Extension Task: Multi-Step Content Processor
import json

# Step 1: Create a translate() function
def translate(text, source_lang='English', target_lang='Spanish'):
    """Translate text from source_lang to target_lang"""
    prompt = f"Translate the following text from {source_lang} to {target_lang}:\n\n{text}\n\nProvide only the translated text."
    try:
        resp = chat.invoke(prompt)
        return resp.content
    except Exception as e:
        return f"Translation error: {str(e)}"

# Step 2: Create a summarize() function
def summarize(text, tone='formal'):
    """Summarize text in one paragraph"""
    prompt = f"Summarize the following text in one short paragraph. Tone: {tone}\n\n{text}"
    try:
        resp = chat.invoke(prompt)
        return resp.content
    except Exception as e:
        return f"Summarization error: {str(e)}"

# Step 3: Create an extract_key_points() function
def extract_key_points(text, max_points=5):
    """Extract key points as a JSON list"""
    prompt = f"""Extract the {max_points} most important points from this text and return ONLY a JSON array of strings.
Format: {{"points": ["point1", "point2", ...]}}

Text:
{text}"""
    try:
        resp = chat.invoke(prompt)
        # Try to parse the JSON response
        parsed = json.loads(resp.content)
        return parsed.get('points', [])
    except json.JSONDecodeError:
        # If JSON parsing fails, try to extract text directly
        return [line.strip() for line in resp.content.split('\n') if line.strip() and line.strip().startswith('-')]
    except Exception as e:
        return [f"Error extracting points: {str(e)}"]

# Step 4: Create a process_content() function that chains everything together
def process_content(text, source_lang='Auto', target_lang='English', tone='formal', max_points=5):
    """
    Multi-step content processor:
    1. Translates to target language (if needed)
    2. Summarizes the content
    3. Extracts key points
    """
    print("🔄 Processing content...\n")
    
    # Translate if not already in target language
    if source_lang != target_lang and source_lang != 'Auto':
        print(f"📝 Step 1: Translating from {source_lang} to {target_lang}...")
        translated = translate(text, source_lang=source_lang, target_lang=target_lang)
        print(f"Translated: {translated[:100]}...\n")
    else:
        translated = text
        print("⏭️  Skipping translation (already in target language)\n")
    
    # Summarize
    print(f"📊 Step 2: Summarizing (tone: {tone})...")
    summary = summarize(translated, tone=tone)
    print(f"Summary: {summary}\n")
    
    # Extract key points
    print(f"🎯 Step 3: Extracting {max_points} key points...")
    points = extract_key_points(summary, max_points=max_points)
    print(f"Key Points: {points}\n")
    
    # Return results as a structured object
    return {
        'original_text': text,
        'translated_text': translated,
        'summary': summary,
        'key_points': points
    }

# Step 5: Test with different texts
print("=" * 60)
print("EXTENSION TASK: Multi-Step Content Processor")
print("=" * 60)

# Test 1: English text
print("\n--- TEST 1: English Article ---\n")
test_text_1 = """
Artificial Intelligence is revolutionizing how we work and live. Machine learning algorithms 
can now recognize images, understand language, and make predictions with remarkable accuracy. 
Deep learning models like transformers have enabled breakthroughs in natural language processing. 
However, ethical concerns about bias, privacy, and job displacement need careful consideration. 
The future of AI depends on developing responsible and transparent systems.
"""

result1 = process_content(test_text_1, source_lang='English', target_lang='English', tone='formal')

print("\n" + "=" * 60)
print("--- TEST 2: Mixed Language Text (French) ---\n")

# Test 2: French text that gets translated
test_text_2 = """
LangChain est un framework puissant qui aide les développeurs à construire des applications
utilisant des modèles de langage. Il fournit des outils pour la gestion des invites et 
l'orchestration des chaînes. Le framework prend en charge plusieurs fournisseurs de modèles.
Avec LangChain, vous pouvez créer des workflows complexes en enchaînant plusieurs opérations.
"""

result2 = process_content(test_text_2, source_lang='French', target_lang='English', tone='casual')

print("\n✅ Extension task complete!")
````
