```python
# Load environment variables from .env file
import os
from dotenv import load_dotenv, find_dotenv

# Load the .env file
load_dotenv(find_dotenv(), override=True)

# Verify the API key is loaded (show first 10 chars only for security)
api_key = os.getenv("OPENAI_API_KEY")
if api_key:
    print(f"✅ API Key loaded: {api_key[:10]}...")
else:
    print("❌ API Key not found! Check your .env file")
```

### LangSmith Tracing
1. Take the LangSmith API KEY from the site
2. Set the three lines in the .env file
- LANGCHAIN_API_KEY=
- LANGCHAIN_TRACING_V2=
- LANGCHAIN_PROJECT=
3. Automatically the tracing ll be present in the LangSmith
