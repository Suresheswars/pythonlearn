## 🛠️ Coding Assignment

**Build a Production-Ready Chatbot with the following features:**

### Requirements:

1. **Initialize LLM**
   - Use `ChatOpenAI` model
   - Set appropriate temperature (0.7 recommended)

2. **Create ChatPromptTemplate**
   - Include a system message that defines the bot's behavior
   - Use `MessagesPlaceholder` for conversation history
   - Include a user input placeholder

3. **Implement Message History**
   - Create a store dictionary to hold `ChatMessageHistory`
   - Use a function to retrieve history (not lambda)
   - Support at least 2 different session IDs

4. **Build RunnableWithMessageHistory Chain**
   - Connect LLM and prompt
   - Configure input and history message keys correctly
   - Ensure proper session management

5. **Run Interactive Chat Loop**
   - Accept user input in a loop
   - Display chatbot responses
   - Allow graceful exit (type 'exit' to quit)
   - Handle errors gracefully

### Code Template:

```python
import os
from dotenv import load_dotenv
import openai

from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override=True)

openai.api_key = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

# Configuration
MODEL_NAME = "gpt-4o-mini"
TEMPERATURE = 0.7
SESSION_ID = "user_session_1"

# Step 1: Initialize LLM
llm = ChatOpenAI(model=MODEL_NAME, temperature=TEMPERATURE)
print(f"✓ LLM initialized: {MODEL_NAME}")

# Step 2: Create prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful, friendly AI assistant. Remember all information the user tells you."),
    MessagesPlaceholder(variable_name="history"),
    ("user", "{input}")
])
print("✓ Prompt template created")

# Step 3: Create history store
store = {SESSION_ID: ChatMessageHistory()}
print("✓ History store created")

# Step 4: Define history retrieval function
def get_history(config):
    session_id = config.get("configurable", {}).get("session_id")
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

# Step 5: Create chatbot with memory
chatbot = RunnableWithMessageHistory(
    prompt | llm,
    get_history,
    input_messages_key="input",
    history_messages_key="history"
)
print("✓ Chatbot with memory created")
print()
print("="*60)
print("🤖 Chatbot Ready!")
print("="*60)
print("Type 'exit' to quit.\n")

# Step 6: Run chat loop
while True:
    user_input = input("You: ").strip()
    if not user_input:
        continue
    if user_input.lower() == 'exit':
        print("\nGoodbye!")
        break
    
    try:
        response = chatbot.invoke(
            {"input": user_input},
            config={"configurable": {"session_id": SESSION_ID}}
        )
        print(f"Bot: {response.content}\n")
    except Exception as e:
        print(f"Error: {str(e)}\n")
```

## ⚠️ Debugging Guide

**If your chatbot isn't working, use this troubleshooting guide:**

| Problem | Likely Cause | Solution |
|---------|--------------|----------|
| Bot doesn't remember name | History not stored | Check that you're using same `session_id` in all calls |
| Bot repeats itself | History includes old responses | This is normal behavior; check if new messages are being added |
| Bot gives generic answers | Missing system prompt | Add system message to ChatPromptTemplate |
| Error about missing API key | OPENAI_API_KEY not set | Create .env file with `OPENAI_API_KEY=your_key_here` |
| Session errors | Store not initialized properly | Make sure store dict has session_id as key |
| Memory not working | Using different session IDs | Keep SESSION_ID consistent throughout your code |
