# Exercise 1 Solution: Extend the Math Server ⭐

## What You Should Have Done
Add `subtract_numbers` and `divide_numbers` tools to the existing `simple_server.py`

---

## Complete simple_server.py with New Tools

````python
# Complete solution for simple_server.py
# This is the EXTENDED version with subtract and divide functions

from mcp.server.fastmcp import FastMCP
import sys
from datetime import datetime

mcp = FastMCP("Math & Time Helper")

# ===== ORIGINAL TOOLS (from session) =====

@mcp.tool()
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together"""
    result = a + b
    print(f"add_numbers: {a} + {b} = {result}", file=sys.stderr)
    return result


@mcp.tool()
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers together"""
    result = a * b
    print(f"multiply_numbers: {a} * {b} = {result}", file=sys.stderr)
    return result


@mcp.tool()
def get_current_time() -> str:
    """Get the current date and time"""
    current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"get_current_time: {current_time}", file=sys.stderr)
    return current_time


# ===== NEW TOOLS (your homework) =====

@mcp.tool()
def subtract_numbers(a: int, b: int) -> int:
    """Subtract b from a and return the result"""
    result = a - b
    print(f"subtract: {a} - {b} = {result}", file=sys.stderr)
    return result


@mcp.tool()
def divide_numbers(a: float, b: float) -> float:
    """Safely divide a by b. Raises error if b is zero"""
    if b == 0:
        raise ValueError("Cannot divide by zero!")
    result = a / b
    print(f"divide: {a} / {b} = {result}", file=sys.stderr)
    return result


if __name__ == "__main__":
    print("Starting Math & Time Helper Server...", file=sys.stderr)
    mcp.run()
````

## Complete simple_client.py with Tests

````python
# Complete solution for simple_client.py
# Tests all 5 tools including the new subtract and divide

import asyncio
from mcp.client.session import ClientSession
from mcp.client.stdio import StdioServerParameters

async def run_client():
    """Test all math and time tools"""
    
    server_params = StdioServerParameters(
        command="python",
        args=["simple_server.py"]
    )
    
    async with ClientSession(server_params) as session:
        # Test 1: Addition
        print("\n=== Testing add_numbers ===")
        result = await session.call_tool(
            "add_numbers",
            arguments={"a": 5, "b": 3}
        )
        print(f"Result: {result.content[0].text}")
        
        # Test 2: Multiplication
        print("\n=== Testing multiply_numbers ===")
        result = await session.call_tool(
            "multiply_numbers",
            arguments={"a": 5, "b": 3}
        )
        print(f"Result: {result.content[0].text}")
        
        # Test 3: Current time
        print("\n=== Testing get_current_time ===")
        result = await session.call_tool(
            "get_current_time",
            arguments={}
        )
        print(f"Result: {result.content[0].text}")
        
        # Test 4: Subtraction (NEW)
        print("\n=== Testing subtract_numbers ===")
        result = await session.call_tool(
            "subtract_numbers",
            arguments={"a": 10, "b": 4}
        )
        print(f"Result: {result.content[0].text}")
        
        # Test 5: Division (NEW)
        print("\n=== Testing divide_numbers ===")
        result = await session.call_tool(
            "divide_numbers",
            arguments={"a": 20, "b": 4}
        )
        print(f"Result: {result.content[0].text}")
        
        # Test 6: Division edge case - by zero
        print("\n=== Testing divide_numbers with zero (error handling) ===")
        try:
            result = await session.call_tool(
                "divide_numbers",
                arguments={"a": 10, "b": 0}
            )
        except Exception as e:
            print(f"Expected error caught: {e}")
        
        print("\n✅ All tests passed!")

if __name__ == "__main__":
    asyncio.run(run_client())
````

# Exercise 2 Solution: String Manipulation Server ⭐⭐

## What You Should Have Done
Create a new `string_server.py` with three string tools, and a client to test them.

---

## Complete string_server.py

````python
# Complete solution for string_server.py

from mcp.server.fastmcp import FastMCP
import sys

mcp = FastMCP("String Helper")


@mcp.tool()
def reverse_string(text: str) -> str:
    """Reverse a string and return it"""
    result = text[::-1]
    print(f"reverse_string: '{text}' -> '{result}'", file=sys.stderr)
    return result


@mcp.tool()
def uppercase_string(text: str) -> str:
    """Convert string to UPPERCASE"""
    result = text.upper()
    print(f"uppercase_string: '{text}' -> '{result}'", file=sys.stderr)
    return result


@mcp.tool()
def count_characters(text: str) -> int:
    """Count the number of characters in a string"""
    result = len(text)
    print(f"count_characters: '{text}' has {result} chars", file=sys.stderr)
    return result


if __name__ == "__main__":
    print("Starting String Helper Server...", file=sys.stderr)
    mcp.run()
````

## Complete string_client.py

````python
# Complete solution for string_client.py

import asyncio
from mcp.client.session import ClientSession
from mcp.client.stdio import StdioServerParameters

async def run_client():
    """Test all string manipulation tools"""
    
    server_params = StdioServerParameters(
        command="python",
        args=["string_server.py"]
    )
    
    async with ClientSession(server_params) as session:
        # Test 1: Reverse string
        print("\n=== Testing reverse_string ===")
        result = await session.call_tool(
            "reverse_string",
            arguments={"text": "hello"}
        )
        print(f"reverse_string('hello') = {result.content[0].text}")
        
        # Test 2: Uppercase
        print("\n=== Testing uppercase_string ===")
        result = await session.call_tool(
            "uppercase_string",
            arguments={"text": "hello world"}
        )
        print(f"uppercase_string('hello world') = {result.content[0].text}")
        
        # Test 3: Count characters
        print("\n=== Testing count_characters ===")
        result = await session.call_tool(
            "count_characters",
            arguments={"text": "hello"}
        )
        print(f"count_characters('hello') = {result.content[0].text}")
        
        # Extra tests
        print("\n=== Extra Tests ===")
        
        # Test with special characters
        result = await session.call_tool(
            "reverse_string",
            arguments={"text": "MCP@2024!"}
        )
        print(f"reverse_string('MCP@2024!') = {result.content[0].text}")
        
        # Test uppercase with numbers
        result = await session.call_tool(
            "uppercase_string",
            arguments={"text": "Python 3.11"}
        )
        print(f"uppercase_string('Python 3.11') = {result.content[0].text}")
        
        # Count long string
        result = await session.call_tool(
            "count_characters",
            arguments={"text": "The Model Context Protocol"}
        )
        print(f"count_characters('The Model Context Protocol') = {result.content[0].text}")
        
        print("\n✅ All string tests passed!")

if __name__ == "__main__":
    asyncio.run(run_client())
````

# Exercise 3 Solution: Temperature Conversion Server ⭐⭐

## What You Should Have Done
Create `temperature_server.py` with three temperature conversion tools and test them with `temperature_client.py`

---

## Complete temperature_server.py

````python
# Complete solution for temperature_server.py

from mcp.server.fastmcp import FastMCP
import sys

mcp = FastMCP("Temperature Converter")


@mcp.tool()
def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert Celsius to Fahrenheit using formula: (C * 9/5) + 32"""
    fahrenheit = (celsius * 9/5) + 32
    print(f"celsius_to_fahrenheit: {celsius}°C = {fahrenheit}°F", file=sys.stderr)
    return fahrenheit


@mcp.tool()
def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """Convert Fahrenheit to Celsius using formula: (F - 32) * 5/9"""
    celsius = (fahrenheit - 32) * 5/9
    print(f"fahrenheit_to_celsius: {fahrenheit}°F = {celsius}°C", file=sys.stderr)
    return celsius


@mcp.tool()
def celsius_to_kelvin(celsius: float) -> float:
    """Convert Celsius to Kelvin using formula: C + 273.15"""
    kelvin = celsius + 273.15
    print(f"celsius_to_kelvin: {celsius}°C = {kelvin}K", file=sys.stderr)
    return kelvin


if __name__ == "__main__":
    print("Starting Temperature Converter Server...", file=sys.stderr)
    mcp.run()
````

## Complete temperature_client.py

````python
# Complete solution for temperature_client.py

import asyncio
from mcp.client.session import ClientSession
from mcp.client.stdio import StdioServerParameters

async def run_client():
    """Test all temperature conversion tools"""
    
    server_params = StdioServerParameters(
        command="python",
        args=["temperature_server.py"]
    )
    
    async with ClientSession(server_params) as session:
        # Test 1: Celsius to Fahrenheit (freezing point)
        print("\n=== Test 1: 0°C to Fahrenheit ===")
        result = await session.call_tool(
            "celsius_to_fahrenheit",
            arguments={"celsius": 0}
        )
        print(f"0°C = {result.content[0].text}°F (Expected: 32.0)")
        
        # Test 2: Celsius to Fahrenheit (boiling point)
        print("\n=== Test 2: 100°C to Fahrenheit ===")
        result = await session.call_tool(
            "celsius_to_fahrenheit",
            arguments={"celsius": 100}
        )
        print(f"100°C = {result.content[0].text}°F (Expected: 212.0)")
        
        # Test 3: Fahrenheit to Celsius (freezing point)
        print("\n=== Test 3: 32°F to Celsius ===")
        result = await session.call_tool(
            "fahrenheit_to_celsius",
            arguments={"fahrenheit": 32}
        )
        print(f"32°F = {result.content[0].text}°C (Expected: 0.0)")
        
        # Test 4: Fahrenheit to Celsius (boiling point)
        print("\n=== Test 4: 212°F to Celsius ===")
        result = await session.call_tool(
            "fahrenheit_to_celsius",
            arguments={"fahrenheit": 212}
        )
        print(f"212°F = {result.content[0].text}°C (Expected: 100.0)")
        
        # Test 5: Celsius to Kelvin (freezing point)
        print("\n=== Test 5: 0°C to Kelvin ===")
        result = await session.call_tool(
            "celsius_to_kelvin",
            arguments={"celsius": 0}
        )
        print(f"0°C = {result.content[0].text}K (Expected: 273.15)")
        
        # Test 6: Celsius to Kelvin (boiling point)
        print("\n=== Test 6: 100°C to Kelvin ===")
        result = await session.call_tool(
            "celsius_to_kelvin",
            arguments={"celsius": 100}
        )
        print(f"100°C = {result.content[0].text}K (Expected: 373.15)")
        
        # Extra tests with different values
        print("\n=== Extra Tests ===")
        
        # Room temperature
        result = await session.call_tool(
            "celsius_to_fahrenheit",
            arguments={"celsius": 20}
        )
        print(f"20°C (room temp) = {result.content[0].text}°F")
        
        # Body temperature
        result = await session.call_tool(
            "celsius_to_fahrenheit",
            arguments={"celsius": 37}
        )
        print(f"37°C (body temp) = {result.content[0].text}°F")
        
        # Negative temperature
        result = await session.call_tool(
            "celsius_to_fahrenheit",
            arguments={"celsius": -40}
        )
        print(f"-40°C = {result.content[0].text}°F (interesting: -40 is same in both scales!)")
        
        print("\n✅ All temperature conversion tests passed!")

if __name__ == "__main__":
    asyncio.run(run_client())
````

## Complete calculator_server.py

````python
# Complete solution for calculator_server.py

from mcp.server.fastmcp import FastMCP
import sys

mcp = FastMCP("Advanced Calculator")


@mcp.tool()
def power(base: float, exponent: float) -> float:
    """Calculate base raised to the power of exponent"""
    result = base ** exponent
    print(f"power({base}, {exponent}) = {result}", file=sys.stderr)
    return result


@mcp.tool()
def square_number(number: float) -> float:
    """Calculate the square of a number"""
    result = number ** 2
    print(f"square_number({number}) = {result}", file=sys.stderr)
    return result


@mcp.tool()
def absolute_value(number: float) -> float:
    """Return the absolute value of a number"""
    result = abs(number)
    print(f"absolute_value({number}) = {result}", file=sys.stderr)
    return result


if __name__ == "__main__":
    print("Starting Advanced Calculator Server...", file=sys.stderr)
    mcp.run()
````

# 📝 Common Student Mistakes & Fixes

## Mistake 1: Forgetting @mcp.tool() decorator
❌ **Wrong:**
```python
def subtract_numbers(a: int, b: int) -> int:
    return a - b
```

✅ **Correct:**
```python
@mcp.tool()
def subtract_numbers(a: int, b: int) -> int:
    return a - b
```

## Mistake 2: Not checking division by zero
❌ **Wrong:**
```python
@mcp.tool()
def divide_numbers(a: float, b: float) -> float:
    return a / b
```

✅ **Correct:**
```python
@mcp.tool()
def divide_numbers(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Cannot divide by zero!")
    return a / b
```

## Mistake 3: Using print() instead of file=sys.stderr
❌ **Wrong:**
```python
print(f"subtract: {a} - {b} = {result}")
```

✅ **Correct:**
```python
print(f"subtract: {a} - {b} = {result}", file=sys.stderr)
```

## Mistake 4: Forgetting type hints
❌ **Wrong:**
```python
@mcp.tool()
def reverse_string(text):
    return text[::-1]
```

✅ **Correct:**
```python
@mcp.tool()
def reverse_string(text: str) -> str:
    """Reverse a string"""
    return text[::-1]
```

## Mistake 5: Wrong temperature formula
❌ **Wrong:**
```python
# This gives wrong results
fahrenheit = celsius * 9/5
```

✅ **Correct:**
```python
# Must add 32
fahrenheit = (celsius * 9/5) + 32
```

## Mistake 6: Not testing edge cases
❌ **Only tests positive numbers:**
```python
result = await session.call_tool("absolute_value", arguments={"number": 5})
```

✅ **Tests multiple cases:**
```python
result = await session.call_tool("absolute_value", arguments={"number": 5})   # positive
result = await session.call_tool("absolute_value", arguments={"number": -5})  # negative
result = await session.call_tool("absolute_value", arguments={"number": 0})   # zero
```

# 💡 Teaching Notes

## Key Concepts to Reinforce

1. **MCP Tool Pattern**: Every tool needs:
   - `@mcp.tool()` decorator
   - Type hints on parameters and return value
   - Docstring explaining what it does
   - Logging to stderr

2. **Server Creation**: Every server needs:
   - `FastMCP("Server Name")`
   - At least one `@mcp.tool()` function
   - `if __name__ == "__main__": mcp.run()`

3. **Client Creation**: Every client needs:
   - `StdioServerParameters` pointing to server.py
   - `async with ClientSession()` context manager
   - `await session.call_tool()` to invoke tools

4. **Error Handling**: Students should:
   - Check for invalid inputs (division by zero)
   - Use try/except in clients
   - Log errors to stderr

## Discussion Questions

1. Why do we log to stderr instead of stdout?
   - **Answer**: stdout is reserved for MCP protocol messages. Logging to stderr keeps debugging info separate.

2. Why do we use type hints in MCP tools?
   - **Answer**: Type hints tell AI agents what kind of data each tool accepts and returns.

3. How could you combine multiple servers into one application?
   - **Answer**: Create a client that calls tools from multiple servers, or create a meta-server that wraps other servers.

4. Why is error handling important in MCP tools?
   - **Answer**: Agents depend on tools working reliably. Proper error messages help agents understand what went wrong.
