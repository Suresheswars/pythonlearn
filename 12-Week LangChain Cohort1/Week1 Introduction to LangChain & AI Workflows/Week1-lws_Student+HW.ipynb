### Q3:
Add simple error handling around the llm.invoke() call to catch missing API key or network errors. Show code that gracefully reports the failure to the user and retries once. Explain how your handler improves robustness.

Submit: code snippet and 3–4 line explanation.

````python
from langchain_openai import ChatOpenAI
from openai import OpenAIError

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.0)

prompt = "Explain LangChain in one line"

def safe_invoke(llm, prompt):
    try:
        return llm.invoke(prompt).content

    except Exception as e:    # catches missing API key, network errors, etc.
        print("⚠️ First attempt failed:", e)
        print("Retrying once...")

        try:
            return llm.invoke(prompt).content
        except Exception as e2:
            print("❌ Second attempt failed:", e2)
            return None

response = safe_invoke(llm, prompt)
print("Final Response:", response)
````

### 🛡️ Robustness Through Error Handling:
**Benefits of this approach:**
1. **Graceful degradation**: App doesn't crash, user gets feedback
2. **Retry logic**: Handles temporary network glitches
3. **User-friendly**: Clear error messages instead of technical stack traces
4. **Production-ready**: Essential for real-world applications

**Common errors this catches:**
- Missing or invalid API key
- Network connectivity issues  
- Rate limiting from OpenAI
- Server timeouts

### Q5:

Combine PromptTemplate with a simple prompt validation: write code that checks input length and refuses to format if the input is empty or too long (e.g., >1000 chars).

Task: Provide the validation function and show example outputs for (a) valid input and (b) invalid input.
Submit: validation code + example outputs.

````python
# 📝 Solution for Q5: Prompt validation with input checks
# =========================================================
# Adding validation ensures robustness and prevents errors

from langchain_core.prompts import PromptTemplate

# Define the prompt template
template = "Summarize the following text: {text}"
prompt = PromptTemplate.from_template(template)

# 🛡️ Validation function with multiple checks
def validate_and_format(prompt_template, text_input):
    """
    Validates input before formatting the prompt.
    
    Args:
        prompt_template: The PromptTemplate object
        text_input: The text to validate
        
    Returns:
        Formatted prompt or error message
    """
    # Check 1: Empty input
    if not text_input or text_input.strip() == "":
        return "❌ Error: Input cannot be empty. Please provide some text."
    
    # Check 2: Input too long (max 1000 characters)
    if len(text_input) > 1000:
        return f"❌ Error: Input too long ({len(text_input)} chars). Maximum allowed: 1000 characters."
    
    # If all checks pass, format the prompt
    return prompt_template.format(text=text_input)

# 📊 Example Test Cases
print("="*70)
print("TEST 1: Valid Input")
print("="*70)
valid_input = "LangChain is a framework for developing AI applications."
result_1 = validate_and_format(prompt, valid_input)
print(f"Input: {valid_input}")
print(f"✅ Output:\n{result_1}\n")

print("="*70)
print("TEST 2: Empty Input")
print("="*70)
empty_input = ""
result_2 = validate_and_format(prompt, empty_input)
print(f"Input: '{empty_input}'")
print(f"{result_2}\n")

print("="*70)
print("TEST 3: Input Too Long (>1000 chars)")
print("="*70)
long_input = "A" * 1500  # Creates a string with 1500 'A's
result_3 = validate_and_format(prompt, long_input)
print(f"Input: String with {len(long_input)} characters")
print(f"{result_3}\n")

# 💡 Benefits of Validation:
print("="*70)
print("📚 Why Validation Matters:")
print("="*70)
print("• Prevents runtime errors from bad input")
print("• Provides clear error messages to users")
print("• Saves API costs by rejecting invalid requests early")
print("• Improves user experience with helpful feedback")
````

##  3️⃣ Chains — Combining Logic Steps

````python
# 🔗 Introduction to Chains - Connecting Steps Together
# =======================================================
# Chains combine prompts and LLMs into reusable workflows

from langchain_openai import OpenAI
from langchain_core.prompts import PromptTemplate

# Step 1: Define your prompt template
prompt = PromptTemplate.from_template(
    "Translate the following text to French:\n\n{text}"
)

# Step 2: Create the LLM (ensure you have OPENAI_API_KEY set)
# Using OpenAI (completion model) instead of ChatOpenAI
llm = OpenAI(model="gpt-3.5-turbo-instruct")

# Step 3: Combine prompt and LLM using the Runnable pipeline
# The | operator creates a RunnableSequence
chain = prompt | llm

# Step 4: Invoke the chain with input data
# The chain automatically formats the prompt then calls the LLM
result = chain.invoke({"text": "LangChain is awesome"})
print(result)
````

### 🔍 Understanding Chains:

**How it works:**
1. **PromptTemplate.from_template()** - Defines how input is formatted into a prompt
2. **prompt | llm** - Creates a RunnableSequence (the chain)
3. **.invoke()** - Runs the full chain and returns the LLM output directly

**The Pipe Operator ( | ):**
- Connects components in sequence
- Output of left side becomes input of right side
- Creates a "pipeline" of operations
- Modern alternative to old `LLMChain` class

### 📝 Key Learning:
Chains are the building blocks of complex AI workflows. They let you:
- Combine multiple steps automatically
- Create reusable workflow patterns
- Build complex logic from simple components
- Maintain clean, readable code

````python
# 📝 Solution for Q3: Combine two models in a chain
# ====================================================
# Use one model for translation, another for polishing

from langchain_openai import OpenAI, ChatOpenAI
from langchain_core.prompts import PromptTemplate

# 🔹 STEP 1: First model - Translation
print("="*70)
print("STEP 1: Translation with gpt-3.5-turbo-instruct")
print("="*70)

translation_prompt = PromptTemplate.from_template(
    "Translate to French: {text}"
)
llm_translator = OpenAI(model="gpt-3.5-turbo-instruct", temperature=0.0)
translation_chain = translation_prompt | llm_translator

input_text = "LangChain makes building AI applications easier and more efficient."
translated_output = translation_chain.invoke({"text": input_text})
print(f"Input: {input_text}")
print(f"Translation: {translated_output.strip()}")

# 🔹 STEP 2: Second model - Polishing
print("\n" + "="*70)
print("STEP 2: Polishing with gpt-4o-mini")
print("="*70)

polishing_prompt = PromptTemplate.from_template(
    "Make this French sentence more elegant and natural:\n\n{text}"
)
llm_polisher = ChatOpenAI(model="gpt-4o-mini", temperature=0.3)
polishing_chain = polishing_prompt | llm_polisher

polished_output = polishing_chain.invoke({"text": translated_output})
print(f"Original translation: {translated_output.strip()}")
print(f"Polished version: {polished_output.content}")

# 📊 COMPARISON
print("\n" + "="*70)
print("📊 COMPARISON:")
print("="*70)
print(f"Model 1 (gpt-3.5-turbo-instruct): {translated_output.strip()}")
print(f"Model 2 (gpt-4o-mini polished):    {polished_output.content}")

print("\n💡 MY PREFERENCE:")
print("The polished version (gpt-4o-mini) looks better because it produces")
print("more natural and elegant French phrasing, making it sound more native.")

# 🎯 Key Learning: You can chain different models for different tasks!
# Translation → Polishing → Formatting → etc.
````

````python
# 🛠️ Introduction to Tools - Extending AI Capabilities
# ======================================================
# Tools allow LLMs to interact with external functions and APIs

from langchain_core.tools import Tool

# Define a simple function
# In real apps, this would call an actual weather API
def get_weather(city: str):
    """Simulate getting weather for a city."""
    return f"The weather in {city} is sunny."

# Wrap the function as a LangChain Tool
tool = Tool(
    name="WeatherTool",  # Unique identifier for the tool
    func=get_weather,  # The actual function to execute
    description="Returns the weather condition for a given city."  # Helps AI understand when to use this tool
)

# Use the tool
print(tool.invoke("Mumbai"))
````

### 📝 What Are Tools?
**Tools** extend what LLMs can do by giving them access to:
- External APIs (weather, stock prices, news)
- Databases (query and retrieve data)
- File systems (read/write files)
- Calculations (math, data processing)
- Web scraping (extract web data)
- Custom business logic

### 🎯 Tool Components:
1. **name**: Unique identifier (used by agents to select tools)
2. **func**: The actual Python function to execute
3. **description**: Explains what the tool does (critical for AI to use it correctly)

### 💡 Why Tools Matter:
LLMs alone have limitations:
- Can't access real-time data
- Can't perform precise calculations
- Can't interact with external systems

Tools solve this by letting LLMs trigger specific functions when needed!

### The preferred interface for tools is now `StructuredTool` or the `@tool` decorator:

````python
# 🆕 Modern Tool Decorator Approach
# ===================================
# The @tool decorator is the preferred way to create tools

from langchain_core.tools import tool

# Use the @tool decorator for cleaner, more Pythonic code
@tool
def get_weather(city: str) -> str:
    """Returns the weather condition for a given city."""
    return f"The weather in {city} is sunny."

# The decorator automatically creates a Tool object
# No need to manually wrap it like before
print(get_weather.invoke("Paris"))
````

### 📝 Tool vs @tool Decorator:
**Old way (Tool class):**
```python
tool = Tool(name="...", func=..., description="...")
```

**New way (@tool decorator):**
```python
@tool
def my_function():
    """Description here"""
    ...
```

**Benefits of @tool:**
- Cleaner, more readable code
- Type hints automatically extracted
- Docstring becomes the description
- Less boilerplate
- More Pythonic

## 5️⃣ Agents — The Brains with Autonomy

## 🧠 AI Workflow Architecture

This diagram shows how different components work together:

```
User Query
   ↓
Prompt Template (structures the input)
   ↓
LLM / Chain (processes and reasons)
   ↓
Tool Calls (if external data/actions needed)
   ↓
Agent Decision (decides next step)
   ↓
Final Response (back to user)
```

### 📝 Key Components:
1. **User Query**: The initial request or question
2. **Prompt Template**: Structures how we ask the AI
3. **LLM/Chain**: The AI reasoning engine
4. **Tool Calls**: External capabilities (APIs, databases, etc.)
5. **Agent Decision**: Logic to determine next steps
6. **Final Response**: The answer delivered to the user

### Coding Section

````python
# 🔄 Multi-Step AI Pipeline Example
# ===================================
# This demonstrates chaining multiple AI operations together

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

# Initialize LLM with balanced creativity
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.5)

# 1️⃣ First Chain: Translation
# Converts non-English text to English
translation_prompt = PromptTemplate.from_template(
    "Translate the following text to English:\n\n{text}"
)
translation_chain = translation_prompt | llm

# 2️⃣ Second Chain: Summarization
# Takes translated text and creates a brief summary
summary_prompt = PromptTemplate.from_template(
    "Summarize the following English text in 3-4 lines:\n\n{translated_text}"
)
summary_chain = summary_prompt | llm

# 3️⃣ Combine chains into a pipeline function
# This orchestrates the multi-step workflow
def pipeline(text):
    """
    Two-step pipeline:
    1. Translate input text to English
    2. Summarize the translated text
    """
    # Step 1: Translate
    translated = translation_chain.invoke({"text": text}).content
    # Step 2: Summarize the translation
    summary = summary_chain.invoke({"translated_text": translated}).content
    return summary

# 4️⃣ Test with Spanish input
input_text = """
El desarrollo de la inteligencia artificial está transformando
la manera en que trabajamos, aprendemos y nos comunicamos en todo el mundo.
"""

# Run the complete pipeline
output = pipeline(input_text)

print("\n🧾 Final Output:")
print(output)
````

### 📝 Multi-Step Pipeline Insights:
**This example demonstrates:**
1. **Sequential processing**: Output of one step feeds into the next
2. **Modular design**: Each chain has a single responsibility
3. **Reusability**: Individual chains can be used independently
4. **Composability**: Chains can be mixed and matched

**Real-world applications:**
- Document processing: Extract → Translate → Summarize → Classify
- Customer support: Detect language → Translate → Analyze sentiment → Route
- Content creation: Research → Draft → Edit → Format

### 💡 Key Takeaway:
Complex AI workflows are built by connecting simple chains together!

## 🔄 LangChain API Updates

### 🧠 Summary of Updates (Old vs New)

| **Old (Deprecated)** | **New (Current)** | **Notes** |
|---------------------|-------------------|-----------|
| `.run()` | `.invoke()` | Standard method to execute chains/models |
| `OpenAI` from langchain | `from langchain_openai import OpenAI` | Moved to separate package |
| `ChatOpenAI` from langchain | `from langchain_openai import ChatOpenAI` | Moved to separate package |
| `LLMChain` | `RunnableSequence` (via `\|` operator) | More flexible, composable |
| Manual error handling | Built-in retry logic | Better fault tolerance |

### 📝 Migration Tips:
1. **Use `.invoke()`** instead of `.run()` for all LLM calls
2. **Import from `langchain_openai`** instead of base `langchain`
3. **Use pipe operator `|`** instead of `LLMChain` when possible
4. **Type hints**: New API has better type safety

### 💡 Why the Changes?
- **Better organization**: Separate packages for different providers
- **Improved performance**: Optimized execution paths
- **Modern patterns**: Embrace Python's functional programming features
- **Extensibility**: Easier to add new features and providers

### “Summarize My Document” — AI Workflow Example

This version demonstrates:

- How user input flows through each stage
- How an LLM chain structures the logic
- Where RAG, memory, and agents could fit in

````python
# 📄 "Summarize My Document" — Complete AI Workflow
# ===================================================
# This example shows a basic but complete AI application

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

# Step 1️⃣: Initialize the LLM
# Lower temperature for more focused, consistent summaries
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.3)

# Step 2️⃣: Create a Prompt Template
# This defines HOW we want the model to process input
prompt = PromptTemplate.from_template(
    "Summarize the following document in simple terms:\n\n{text}"
)

# Step 3️⃣: Create a Runnable pipeline (Prompt → LLM)
# The chain connects template formatting with LLM processing
chain = prompt | llm

# Step 4️⃣: User query (simulating user input)
# In a real app, this would come from a file upload or text input
user_query = """
Machine learning models improve automatically through experience.
They are used in recommendation systems, speech recognition, and autonomous vehicles.
"""

# Step 5️⃣: Invoke the chain
# This formats the prompt and gets the AI summary
response = chain.invoke({"text": user_query})

print("\n🧾 Final Output:")
print(response.content)
````
