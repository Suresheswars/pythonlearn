## 1️⃣ LLMs (Language Models)

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-3.5-turbo")
response = llm.invoke("Explain LangChain in 1 line")
print(response)
```

## 2️⃣ Prompts — Giving Context to LLMs

```python
from langchain_core.prompts import PromptTemplate

template = "Translate the following sentence to French: {text}"
prompt = PromptTemplate.from_template(template)
print(prompt.format(text="I love building AI apps!"))
```

##  3️⃣ Chains — Combining Logic Steps

### Explanation:
1. PromptTemplate.from_template() defines how input is formatted.
2. prompt | llm creates a RunnableSequence, equivalent to the old LLMChain.
3. .invoke() runs the full chain and returns the LLM output directly.
