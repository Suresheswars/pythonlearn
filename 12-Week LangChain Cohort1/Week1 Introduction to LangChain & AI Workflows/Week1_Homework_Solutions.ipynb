## Question 3: Error Handling

**Task:** Add simple error handling around the llm.invoke() call to catch missing API key or network errors. Show code that gracefully reports the failure to the user and retries once.

```python
# Solution for Question 3
# =======================

from langchain_openai import ChatOpenAI
from openai import OpenAIError

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.0)
prompt = "Explain LangChain in one line"

def safe_invoke(llm, prompt):
    """
    Safely invoke LLM with error handling and retry logic.
    
    Args:
        llm: The LLM instance
        prompt: The prompt to send
        
    Returns:
        str: The response content or None if both attempts fail
    """
    try:
        # First attempt
        return llm.invoke(prompt).content
        
    except Exception as e:
        # Catches missing API key, network errors, rate limits, etc.
        print("⚠️ First attempt failed:", str(e)[:100])  # Truncate long errors
        print("🔄 Retrying once...")
        
        try:
            # Second attempt
            return llm.invoke(prompt).content
            
        except Exception as e2:
            print("❌ Second attempt failed:", str(e2)[:100])
            print("💡 Please check your API key and network connection.")
            return None

# Test the function
response = safe_invoke(llm, prompt)

if response:
    print("\n✅ Final Response:")
    print(response)
else:
    print("\n❌ Unable to get response after retry.")
```

### 📝 Explanation - How This Improves Robustness:

**1. Graceful Error Handling:**
   - Doesn't crash the application on errors
   - Provides user-friendly error messages instead of technical stack traces

**2. Automatic Retry Logic:**
   - Handles temporary network glitches automatically
   - One retry is often enough for transient errors

**3. Informative Feedback:**
   - Users know what went wrong and what the system is doing
   - Clear indicators of success/failure states

**4. Production-Ready:**
   - Returns None on failure, allowing calling code to handle it
   - Prevents cascading failures in larger applications

This pattern is essential for building reliable AI applications!

**Task:** Show what happens when a required variable is missing. Explain why the error occurs and how to avoid it.

```python
# Solution for Prompts Q3
# ========================

from langchain_core.prompts import PromptTemplate

# Using the original translation template
template = "Translate the following sentence to French: {text}"
prompt = PromptTemplate.from_template(template)

print("="*70)
print("MISSING VARIABLE ERROR DEMONSTRATION")
print("="*70)
print("\n🔴 Attempting to format without required 'text' variable:\n")

try:
    # This will cause an error
    result = prompt.format()
    print(result)
    
except KeyError as e:
    print(f"❌ Error occurred: KeyError - {e}")
    print("\n📚 Explanation:")
    print("   • The error occurs because the template expects a 'text' variable")
    print("   • We called format() without providing text=...")
    print("   • Python raises a KeyError when a required key is missing")
    
print("\n✅ How to avoid this error:")
print("   1. Always provide all required variables in format()")
print("   2. Use clear variable names to remember what's needed")
print("   3. Add input validation before calling format()")

print("\n💡 Correct Usage:")
correct_result = prompt.format(text="I love building AI apps!")
print(correct_result)
```

### 📝 Explanation:

- Templates have placeholders that MUST be filled
- Python's string formatting raises `KeyError` for missing keys
- The template can't be completed without all variables

**How to avoid:**
1. Always check template requirements before formatting
2. Use try-except blocks for user input scenarios
3. Validate inputs before passing to format()
4. Keep track of required variables in documentation

## Question 5: Input Validation

**Task:** Write validation code that checks input length and refuses to format if empty or too long (>1000 chars).

```python
# Solution for Prompts Q5
# ========================

from langchain_core.prompts import PromptTemplate

# Define the prompt template
template = "Summarize the following text: {text}"
prompt = PromptTemplate.from_template(template)

# Validation function
def validate_and_format(prompt_template, text_input):
    """
    Validates input before formatting the prompt.
    
    Args:
        prompt_template: The PromptTemplate object
        text_input: The text to validate
        
    Returns:
        Formatted prompt or error message
    """
    # Check 1: Empty input
    if not text_input or text_input.strip() == "":
        return "❌ Error: Input cannot be empty. Please provide some text."
    
    # Check 2: Input too long
    if len(text_input) > 1000:
        return f"❌ Error: Input too long ({len(text_input)} chars). Maximum: 1000 characters."
    
    # If all checks pass, format the prompt
    return prompt_template.format(text=text_input)

# TEST CASES
print("="*70)
print("INPUT VALIDATION TESTING")
print("="*70)

# Test 1: Valid input
print("\n📊 TEST 1: Valid Input")
print("="*70)
valid_input = "LangChain is a framework for developing AI applications."
result_1 = validate_and_format(prompt, valid_input)
print(f"Input: {valid_input}")
print(f"✅ Output:\n{result_1}")

# Test 2: Empty input
print("\n📊 TEST 2: Empty Input")
print("="*70)
empty_input = ""
result_2 = validate_and_format(prompt, empty_input)
print(f"Input: '{empty_input}'")
print(result_2)

# Test 3: Too long input
print("\n📊 TEST 3: Input Too Long (>1000 chars)")
print("="*70)
long_input = "A" * 1500  # Creates 1500 'A's
result_3 = validate_and_format(prompt, long_input)
print(f"Input: String with {len(long_input)} characters")
print(result_3)
```

### 📝 Why Validation Matters:

**Production Benefits:**
1. **Prevents errors**: Catches issues before API calls
2. **Saves money**: Avoids wasted API tokens on invalid requests
3. **Better UX**: Clear error messages help users fix problems
4. **Security**: Prevents malicious or malformed inputs
5. **Reliability**: Makes applications more robust

**Best Practices:**
- Validate early (before expensive operations)
- Provide helpful error messages
- Set reasonable limits
- Log validation failures for monitoring
- Consider edge cases

# 🔗 Section 3: Chains — Combining Logic Steps

## Question 1: Chain Execution

**Task:** Run the chain and print both full output and text only.

```python
# Solution for Chains Q1
# =======================

from langchain_openai import OpenAI
from langchain_core.prompts import PromptTemplate

# Define prompt template
prompt = PromptTemplate.from_template(
    "Translate the following text to French:\n\n{text}"
)

# Create the LLM
llm = OpenAI(model="gpt-3.5-turbo-instruct", temperature=0.0)

# Combine using the Runnable pipeline
chain = prompt | llm

# Invoke the chain
result = chain.invoke({"text": "LangChain is awesome"})

# Display results
print("="*70)
print("FULL OUTPUT (raw response):")
print("="*70)
print(result)
print(f"\nType: {type(result)}")

print("\n" + "="*70)
print("TEXT ONLY (clean output):")
print("="*70)
print(result.strip())

print("\n💡 Note: For OpenAI completion models, the result is already a string.")
print("   For ChatOpenAI, you would use: result.content")
```

### 📝 Key Learning:
- The `|` operator creates a chain (RunnableSequence)
- Output type depends on the model used
- OpenAI (completion): returns string directly
- ChatOpenAI: returns object, use `.content` for text

```python
# Solution for Tools Q1
# ======================

from langchain_core.tools import Tool

def get_weather(city: str) -> str:
    """
    Returns detailed weather information for a given city.
    
    In a real application, this would call a weather API.
    For this demo, we return realistic-looking simulated data.
    """
    # Simulated weather database
    weather_data = {
        "Mumbai": {"temp": 32, "condition": "Partly cloudy with high humidity"},
        "Paris": {"temp": 18, "condition": "Light rain expected in evening"},
        "New York": {"temp": 22, "condition": "Clear skies, pleasant weather"},
        "Tokyo": {"temp": 25, "condition": "Sunny with gentle breeze"},
        "London": {"temp": 15, "condition": "Overcast with occasional drizzle"}
    }
    
    # Get weather for city (default if not found)
    weather = weather_data.get(city, {"temp": 20, "condition": "Data not available"})
    
    # Return formatted response
    return f"""🌡️ Weather in {city}:
   Temperature: {weather['temp']}°C
   Condition: {weather['condition']}"""

# Create the Tool
tool = Tool(
    name="WeatherTool",
    func=get_weather,
    description="Returns detailed weather information (temperature and condition) for a given city."
)

# Test the enhanced tool
print("="*70)
print("TESTING ENHANCED WEATHER TOOL")
print("="*70)

test_cities = ["Mumbai", "Paris", "Berlin"]
for city in test_cities:
    print(f"\n{tool.invoke(city)}")
```

## Question 2: Input Validation

**Task:** Add validation to check if city is empty.

```python
# Solution for Tools Q2
# ======================

from langchain_core.tools import Tool

def get_weather(city: str) -> str:
    """
    Returns weather information with input validation.
    """
    # VALIDATION: Check if city is empty or whitespace
    if not city or city.strip() == "":
        return "⚠️ Please provide a city name."
    
    # Weather data
    weather_data = {
        "Mumbai": "🌡️ 32°C, Partly cloudy with high humidity",
        "Paris": "🌡️ 18°C, Light rain expected",
        "New York": "🌡️ 22°C, Clear skies",
        "Tokyo": "🌡️ 25°C, Sunny with gentle breeze"
    }
    
    # Return weather or default
    if city in weather_data:
        return f"Weather in {city}: {weather_data[city]}"
    else:
        return f"Weather in {city}: 🌡️ 20°C, Sunny (default forecast)"

# Create the Tool
tool = Tool(
    name="WeatherTool",
    func=get_weather,
    description="Returns weather information for a given city with input validation."
)

# DEMONSTRATION
print("="*70)
print("INPUT VALIDATION DEMONSTRATION")
print("="*70)

# Test 1: Valid city
print("\n📊 TEST 1: Valid City Name")
print("="*70)
result_valid = tool.invoke("Mumbai")
print(f"Input: 'Mumbai'")
print(f"Output: {result_valid}")

# Test 2: Empty city
print("\n📊 TEST 2: Empty City Name")
print("="*70)
result_empty = tool.invoke("")
print(f"Input: '' (empty string)")
print(f"Output: {result_empty}")
```

## Question 3: Test Harness

**Task:** Create automated tests for multiple cities.

```python
# Solution for Tools Q3
# ======================

from langchain_core.tools import Tool

def get_weather(city: str) -> str:
    """Returns weather information for a city."""
    if not city or city.strip() == "":
        return "⚠️ Please provide a city name."
    
    weather_data = {
        "Mumbai": "🌡️ 32°C, Partly cloudy",
        "Paris": "🌡️ 18°C, Light rain",
        "Nowhere": "🌡️ Data not available for this location"
    }
    
    return weather_data.get(city, f"🌡️ 20°C, Sunny (forecast for {city})")

# Create the Tool
tool = Tool(
    name="WeatherTool",
    func=get_weather,
    description="Returns weather condition for a given city."
)

# AUTOMATED TEST HARNESS
print("="*70)
print("🧪 AUTOMATED TEST HARNESS")
print("="*70)

# Test data
test_cities = ["Mumbai", "Paris", "Nowhere"]

# Collect results
results = []

print("\n📋 Running tests...\n")
for city in test_cities:
    result = tool.invoke(city)
    results.append({
        "city": city,
        "weather": result
    })
    print(f"✓ Testing '{city}'")
    print(f"  → {result}\n")

# Display collected results
print("="*70)
print("📊 COLLECTED RESULTS (List Format):")
print("="*70)
print(results)

# Display as formatted table
print("\n" + "="*70)
print("📈 RESULTS TABLE:")
print("="*70)
print(f"{'City':<15} {'Weather Information':<50}")
print("-"*70)
for item in results:
    print(f"{item['city']:<15} {item['weather']:<50}")
```

### 📝 Test Harness Benefits:

**Why automated testing?**
1. **Saves time**: Test multiple scenarios instantly
2. **Consistency**: Same tests every time
3. **Catch regressions**: Detect when changes break things
4. **Documentation**: Tests show how tools should work
5. **Confidence**: Know your code works before deploying

**Best practices:**
- Test normal cases
- Test edge cases (empty, null, extreme values)
- Test error conditions
- Automate repetitive tests
- Keep tests simple and focused
