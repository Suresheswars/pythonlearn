   - `LANGCHAIN_API_KEY` = LANGSMITH_API_KEY

3. Initialize a ChatOpenAI LLM with:
   - model: "gpt-3.5-turbo"
   - temperature: 0.7
   - api_key: OPENAI_API_KEY

4. Create a prompt template with text: "You are a helpful AI assistant. Answer this question: {question}"

5. Create a chain using LCEL (pipe operator `|`) combining prompt and llm

6. Run the chain with question: "What is the capital of France?"

7. Print the answer and handle any errors

### Expected Output:
```
🔄 Running first traced chain execution...
Answer: [LLM response]
```

````python
# TODO: Import modules, set env vars, initialize LLM, create prompt, create chain, run chain

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langsmith import traceable
import os

# os.environ["LANGCHAIN_TRACING_V2"] = "true"
# os.environ["LANGCHAIN_PROJECT"] = PROJECT_NAME
# os.environ["LANGCHAIN_API_KEY"] = LANGSMITH_API_KEY
# os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY

# Initialize LLM:
llm = ChatOpenAI(
    model = "gpt-3.5-turbo",
    temperature = 0.7,
    api_key = OPENAI_API_KEY
)

# Create a simple prompt:
prompt = ChatPromptTemplate.from_template(
    "you are a helpful assistant. Answer this question {question}."
)

# Pipeoperator from the LCEL:
chain = prompt | llm

result = chain.invoke({"question": "What is the capital of India?"})

print(f"Answer: {result.content}")
````

## Summary of LangSmith Components

| Component | Purpose | When to Use |
|-----------|---------|----------|
| **Client** | Connect to LangSmith API | Every application |
| **Projects** | Organize runs and experiments | Separate use cases |
| **Tracing** | Capture execution logs | Debug chains and agents |
| **Custom Decorators** | Trace custom functions | Track custom code |
| **Datasets** | Store test cases | Evaluation and testing |
| **Evaluators** | Score run quality | Measure performance |
| **Feedback** | Collect user ratings | Improve models |
| **Hub** | Share prompts | Team collaboration |
| **Run Analysis** | Monitor performance | Production monitoring |
| **Error Handling** | Capture exceptions | Debugging production |
