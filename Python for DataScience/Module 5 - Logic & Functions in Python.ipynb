## Lambda Functions (Anonymous Functions)

````python
# Traditional function approach
# Define a function that squares a number

def square(x):
    """Returns the square of a number"""
    return x * x

result = square(5)
print(f"Square of 5 using traditional function: {result}")
````

````python
square_lambda = lambda x: x * x
````

````python
nums = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, nums))

print(squares)
````

````python
# map() applies a function to every item in a sequence
# Useful for transforming lists

nums = [1, 2, 3, 4]
# Apply square lambda to each number in the list
squares = list(map(lambda x: x * x, nums))

print(f"Original list: {nums}")
print(f"Squared list: {squares}")
````

## Variable Scope: Local vs Global

````python
# filter() keeps only items that satisfy a condition
# Useful for selecting elements from a sequence

is_even = lambda x: x % 2 == 0

nums = [1, 2, 3, 4, 5, 6]
# Keep only even numbers
evens = list(filter(is_even, nums))

print(f"Original list: {nums}")
print(f"Even numbers only: {evens}")
````

### Mini Project: Build a Calculator using Functions

#### Features:

- Add, Subtract, Multiply, Divide

- Each operation in a separate function

- Take input from user

- Use while True to keep running until user exits

````python
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b): return a / b if b != 0 else "Cannot divide by zero"

while True:
    print("\n1. Add | 2. Subtract | 3. Multiply | 4. Divide | 5. Exit")
    choice = input("Enter choice: ")

    if choice == "5":
        break

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == "1":
        print("Result:", add(a, b))
    elif choice == "2":
        print("Result:", subtract(a, b))
    elif choice == "3":
        print("Result:", multiply(a, b))
    elif choice == "4":
        print("Result:", divide(a, b))
    else:
        print("Invalid choice.")
````

`filter()` – Filter items using a condition

`filter(function, iterable)`

````python
def is_even(x):
    return x % 2 == 0

nums = [1, 2, 3, 4, 5]

list(filter(is_even, nums))
````

````python
# reduce() - Combine all items in a sequence into a single value
# Requires importing from functools

from functools import reduce

numbers = [1, 2, 3, 4]
# Reduce: (1+2) = 3, (3+3) = 6, (6+4) = 10
total = reduce(lambda a, b: a + b, numbers)

print(f"Numbers: {numbers}")
print(f"Sum using reduce: {total}")
````

## Function-Based Calculator

### Features:
- Add, Subtract, Multiply, Divide

- Power, Modulus

- Use functions for each operation

- Use map() to collect multiple inputs

````python
# Advanced Calculator Functions
# Modular design: each operation is a separate, reusable function

def add(a, b):
    """Add two numbers"""
    return a + b

def subtract(a, b):
    """Subtract two numbers"""
    return a - b

def multiply(a, b):
    """Multiply two numbers"""
    return a * b

def divide(a, b):
    """Divide with error handling for zero division"""
    return a / b if b != 0 else "Error: Division by zero"

def power(a, b):
    """Calculate power (a to the power b)"""
    return a ** b

def modulus(a, b):
    """Get remainder of division"""
    return a % b

def get_inputs():
    """Get two numbers from user with validation"""
    try:
        nums = input("Enter two numbers separated by space: ")
        a, b = map(float, nums.split())
        return a, b
    except ValueError:
        print("❌ Please enter valid numbers.")
        return get_inputs()  # Ask again if invalid input
````

````python
while True:
    print("\nCALCULATOR MENU")
    print("1. Add\n2. Subtract\n3. Multiply\n4. Divide\n5. Power\n6. Modulus\n7. Exit")

    choice = input("Choose an option (1–7): ")

    if choice == "7":
        print("Thank you for using the calculator.")
        break

    a, b = get_inputs()

    if choice == "1":
        print("Result:", add(a, b))
    elif choice == "2":
        print("Result:", subtract(a, b))
    elif choice == "3":
        print("Result:", multiply(a, b))
    elif choice == "4":
        print("Result:", divide(a, b))
    elif choice == "5":
        print("Result:", power(a, b))
    elif choice == "6":
        print("Result:", modulus(a, b))
    else:
        print("Invalid choice. Try again.")
````

### Modular Function-Based Calculator

````python
# Advanced Calculator with History Tracking
# Demonstrates functions, loops, and data structures working together

def show_menu():
    """Display calculator menu options"""
    print("\n" + "="*40)
    print("📋 CALCULATOR MENU")
    print("="*40)
    print("1. Add (+)")
    print("2. Subtract (-)")
    print("3. Multiply (×)")
    print("4. Divide (÷)")
    print("5. Power (**)")
    print("6. Modulus (%)")
    print("7. View History")
    print("8. Exit")
    print("="*40)

def calculator():
    """Main calculator function"""
    history = []  # Store all calculations
    
    while True:
        show_menu()
        choice = input("Choose an option (1-8): ").strip()
        
        if choice == "8":
            print("\n✓ Thank you for using the calculator!")
            break
            
        elif choice == '7':
            if history:
                print("\n📝 Previous Calculations:")
                for i, entry in enumerate(history, 1):
                    print(f"  {i}. {entry}")
            else:
                print("\n📝 No calculations yet.")
            continue
        
        if choice not in {"1", "2", "3", "4", "5", "6"}:
            print("\n❌ Invalid choice! Please try again.")
            continue
            
        a, b = get_inputs()
        
        # Perform calculation based on choice
        if choice == "1":
            result = add(a, b)
            op = "+"
        elif choice == "2":
            result = subtract(a, b)
            op = "-"
        elif choice == "3":
            result = multiply(a, b)
            op = "×"
        elif choice == "4":
            result = divide(a, b)
            op = "÷"
        elif choice == "5":
            result = power(a, b)
            op = "**"
        elif choice == "6":
            result = modulus(a, b)
            op = "%"
        
        print(f"\n📊 Result: {a} {op} {b} = {result}")
        history.append(f"{a} {op} {b} = {result}")

# Uncomment to run the calculator
# calculator()
````

# 🎓 MODULE 5 SESSION SUMMARY

## ✅ What You've Learned Today

### 1️⃣ Logical Operators
- **`and`**: Both conditions must be True
- **`or`**: At least one condition must be True
- **`not`**: Inverts the boolean value
- Used for complex decision-making in conditionals

### 2️⃣ Functions - The Building Blocks
- **Defining functions** with `def` keyword
- **Parameters**: Inputs that functions accept
- **Return values**: Outputs that functions produce
- **Function calls**: How to use functions in your code

### 3️⃣ Types of Parameters
- **Positional parameters**: Order matters (e.g., `greet(name, age)`)
- **Default parameters**: Have default values (e.g., `greet(name, age=18)`)
- **Keyword arguments**: Specify parameter names (e.g., `greet(name="John", age=25)`)
- **`*args`**: Accept unlimited positional arguments
- **`**kwargs`**: Accept unlimited keyword arguments

### 4️⃣ Variable Scope
- **Local scope**: Variables inside functions (only accessible within that function)
- **Global scope**: Variables outside functions (accessible everywhere)
- **`global` keyword**: Modify global variables from inside functions

### 5️⃣ Lambda Functions
- **Anonymous functions**: Quick one-line functions
- **Syntax**: `lambda parameters: expression`
- **Use cases**: Short operations, sorting, filtering
- **Example**: `lambda x: x * 2`

### 6️⃣ Docstrings & Documentation
- **Triple quotes**: `"""Function description"""`
- **Purpose**: Explain what your function does, parameters, and return values
- **Access docstrings**: Using `help(function_name)` or `function_name.__doc__`

---

## 🎯 Key Best Practices

1. **Always use meaningful function names**
   - ✅ `calculate_total_price()`
   - ❌ `func1()`

2. **Write docstrings for every function**
   - Helps others (and future you!) understand your code
   - Makes debugging easier

3. **Keep functions focused**
   - One function = One task
   - If it's doing too much, split it into smaller functions

4. **Use default parameters wisely**
   - Provide sensible defaults
   - Make functions easier to use

5. **Validate inputs**
   - Check for invalid data before processing
   - Raise helpful error messages

6. **Return values instead of printing**
   - Makes functions reusable
   - More flexible for different use cases

## 💡 Common Patterns You Should Know

```python
# Pattern 1: Input validation
def safe_divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

# Pattern 2: Using *args for flexible inputs
def sum_all(*numbers):
    return sum(numbers)

# Pattern 3: Using **kwargs for options
def create_profile(name, **details):
    profile = {'name': name}
    profile.update(details)
    return profile

# Pattern 4: Lambda for quick operations
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
```
