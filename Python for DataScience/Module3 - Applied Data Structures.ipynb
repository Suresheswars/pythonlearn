## Student Record Plan

### 1) Represent a Student
Use a dictionary so fields are self-describing.

```python
# One student record (dictionary)
student = {"name": "Aarav", "roll": 101, "marks": 88}

# Storage for all records (list of dicts)
records = []
records.append(student)
print(records)
```

### 2) Add a Student

```python
# Gather inputs with simple validation
while True:
    try:
        name = input("Enter Name: ").strip().title()
        roll = int(input("Enter roll number: "))
        marks = int(input("Enter marks (0-100): "))
        if not 0 <= marks <= 100:
            print("Marks must be between 0 and 100. Try again.\n")
            continue
        break
    except ValueError:
        print("Please enter numbers for roll and marks. Try again.\n")
```

```python
print(f"Captured -> Name: {name}, Roll: {roll}, Marks: {marks}")
```

```python
# Build and store the student record
student = {"name": name, "roll": roll, "marks": marks}
records.append(student)
print(records)
```

```python
print(records)
```

## View all students

```python
for s in records:
    print(s)
```

### Search by Name
Use a comprehension to filter matching records.

```python
if not records:
    print("No records yet. Add students first.")
else:
    search = input("Enter name to search: ").strip().title()
    # Filter list for matching names
    found = [s for s in records if s["name"].title() == search]
    if found:
        print(found[0])
    else:
        print("Student not found.")
```

### Remove Student by Roll
Filter out the matching roll number.

```python
if not records:
    print("No records to remove.")
else:
    roll_to_remove = int(input("Enter roll number to remove: "))
    before = len(records)
    records = [s for s in records if s['roll'] != roll_to_remove]
    if len(records) < before:
        print("Removed.")
    else:
        print("Roll not found.")
```

## Add Menu Loop
Tie all operations together in a simple CLI.

```python
# Ensure records exists even if you jump to this cell
if "records" not in locals():
    records = []

def add_student():
    while True:
        try:
            name = input("Enter Name: ").strip().title()
            roll = int(input("Enter roll number: "))
            marks = int(input("Enter marks (0-100): "))
            if not 0 <= marks <= 100:
                print("Marks must be between 0 and 100. Try again.\n")
                continue
            records.append({"name": name, "roll": roll, "marks": marks})
            print("Added.")
            break
        except ValueError:
            print("Please enter numbers for roll and marks. Try again.\n")

def view_students():
    if not records:
        print("No records yet.")
    else:
        for s in records:
            print(s)

def search_student():
    if not records:
        print("No records yet.")
        return
    search = input("Enter name to search: ").strip().title()
    found = [s for s in records if s["name"].title() == search]
    print(found[0] if found else "Student not found.")

def remove_student():
    if not records:
        print("No records to remove.")
        return
    roll_to_remove = int(input("Enter roll number to remove: "))
    before = len(records)
    records[:] = [s for s in records if s['roll'] != roll_to_remove]
    print("Removed." if len(records) < before else "Roll not found.")

def show_average():
    avg = sum(s["marks"] for s in records) / len(records) if records else 0
    print(f"Average: {avg:.2f}")

def show_topper():
    top = max(records, key=lambda s: s["marks"]) if records else None
    print(top if top else "No records yet.")

while True:
    print("1. Add | 2. View | 3. Search | 4. Remove | 5. Avg | 6. Top | 7. Exit")
    choice = input("Enter Choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        view_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        remove_student()
    elif choice == "5":
        show_average()
    elif choice == "6":
        show_topper()
    elif choice == "7":
        break
    else:
        print("Invalid choice")
```
