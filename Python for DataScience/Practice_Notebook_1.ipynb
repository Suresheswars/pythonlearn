## 🎯 Quick Markdown Reference

| Element | Syntax | Example |
|---------|--------|----------|
| Heading 1 | `# Text` | # Heading |
| Heading 2 | `## Text` | ## Subheading |
| Bold | `**text**` | **bold** |
| Italic | `*text*` | *italic* |
| Strikethrough | `~~text~~` | ~~crossed~~ |
| Inline Code | `` `code` `` | `print()` |
| Code Block | ``` ```python ``` | See examples above |
| Bullet List | `- item` | - Item |
| Numbered List | `1. item` | 1. Item |
| Quote | `> text` | > Quote |
| Link | `[text](url)` | [Link](url) |
| Image | `![alt](url)` | ![img](url) |

---

## 💡 Pro Tips

1. **Use Markdown for documentation**: Always document your code with clear explanations
2. **Combine formatting**: Mix **bold**, *italic*, and `code` for emphasis
3. **Preview mode**: Double-click a markdown cell to edit, run it to render
4. **Organize with headings**: Structure your notebooks like a document
5. **Visual hierarchy**: Use headings, lists, and quotes to make notebooks scannable
