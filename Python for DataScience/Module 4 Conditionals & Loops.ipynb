# Conditionals – if, elif, else

#### What Are Conditionals?

Conditional statements allow you to __execute different blocks of code based on certain conditions__ (True/False values).

- `if` – basic condition check

- `elif` – check another condition if the previous one failed

- `else` – run this block if all previous conditions failed

if condition:
    # Code block if condition is True
elif another_condition:
    #Code block if this condition is True
else:
    #Code block if all above conditions are False

### Basic Example: Voting Eligibility

```python
# Voting Example - with input validation
# This demonstrates: input validation + conditional logic

while True:
    try:
        age = int(input("Enter your age (0-120): "))
        if 0 <= age <= 120:
            break  # Valid input, exit loop
        else:
            print("❌ Please enter a valid age between 0 and 120.")
    except ValueError:
        print("❌ Please enter a valid number (no letters).")

# Now check eligibility
if age >= 18:
    print("✓ You are eligible to vote!")
else:
    print("✗ You are not eligible to vote yet. Come back in a few years!")
```

### Grading System

```python
# Grading System - with input validation
# This demonstrates: nested conditionals + input validation

while True:
    try:
        marks = int(input("Enter your marks (0-100): "))
        if 0 <= marks <= 100:
            break  # Valid input, exit loop
        else:
            print("❌ Please enter marks between 0 and 100.")
    except ValueError:
        print("❌ Please enter a valid number.")

# Assign grade based on marks
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 35:
    grade = "C"
else:
    grade = "F (Fail)"

print(f"Your Grade: {grade}")
```

### Nested `if` Statements

```python
# Nested If Statements - with validation
# This demonstrates: nested logic + input validation

name = input("Enter your name: ").strip()  # strip() removes extra spaces

while True:
    try:
        age = int(input("Enter your age: "))
        if age > 0 and age < 150:
            break
        else:
            print("❌ Please enter a realistic age.")
    except ValueError:
        print("❌ Please enter a number.")

# Check eligibility with nested conditions
if age >= 18:
    if name.lower() == "aarav":
        print(f"🌟 Special access granted to {name}!")
    else:
        print(f"✓ Standard access granted to {name}")
else:
    print(f"✗ Access Denied! {name}, you must be 18+")
```

### Simple Login System

```python
# Simple Login System with 3 attempts
# This demonstrates: for loop + break/else + input validation

correct_user = "admin"
correct_pass = "1234"

for attempt in range(1, 4):  # 1, 2, 3 (3 attempts total)
    username = input(f"\nAttempt {attempt}: Username: ").strip()
    password = input("Password: ").strip()

    if username == correct_user and password == correct_pass:
        print("🔓 Login Successful! Welcome Admin.")
        break
    else:
        remaining = 3 - attempt
        if remaining > 0:
            print(f"❌ Incorrect credentials. {remaining} attempt(s) remaining.")
else:
    # This 'else' runs only if loop completes WITHOUT 'break'
    print("\n🔒 Account locked. Too many failed attempts. Contact support.")
```

### What is a for loop?

A `for` loop is used to __iterate__ over a sequence like a list, string, range, or dictionary and repeat a block of code __for each item__.

for variable in sequence:
    ## code block

### Example1: Print Numbers 1to5

```python
for i in range(1, 6):
    print(i)
```

### Example 2: Loop through a list of names

```python
students = ['Aarav', 'Divya', 'Meera', 'Sanjay']

for name in students:
    print("Hello", name)
```

### Example3: Loop through characters in a string

```python
for i in "Python":
    print(i)
```

### Loop through Dictionary

```python
marks = {"Aarav": 85, "Divya": 90, "Meera": 78}

for name, score in marks.items():
    print(name, "scored", score)
```

### Loop with `if` condition inside:

```python
numbers = [10, 25, 30, 42, 55]

for num in numbers:
    if num % 2 == 0:
        print(num, "is even")
```

### Attendence Sheet: Mini Project

```python
students = ['Aarav', 'Divya', 'Meera', 'Sanjay']
present = []

for name in students:
    status = input(f"Is {name} present? (y/n): ")
    if status.lower() == 'y':
        present.append(name)

print("Present Students:", present)
```

## `while` loop

while condition:

### Example 1: Basic Calculator

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

### Countdown

```python
count = 10

while count > 0:
    print("Countdown:", count)
    count -= 1
```
