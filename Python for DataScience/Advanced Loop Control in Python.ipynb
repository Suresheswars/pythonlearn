### `break` and `continue`

`break` → Exits the loop immediately

````python
# Example: Using break to exit loop early
# Scenario: Stop searching once we find the target student

for student in ["Aarav", "Divya", "Meera", "Ravi"]:
    if student == "Meera":
        print("Found Meera, stop search")
        break  # Exit the loop immediately - remaining names won't be checked
    print("Checking", student)

# Output: Checks Aarav, Divya, then finds Meera and stops (Ravi is never checked)
````

`continue` → Skips the current iteration

````python
# Example: Using continue to skip specific iterations
# Scenario: Process all names except "Divya"

for name in ["Aarav", "Divya", "Meera", "Ravi"]:
    if name == "Divya":
        continue  # Skip this iteration - go to next name
    print("Processing:", name)

# Output: Processes Aarav, skips Divya, then processes Meera and Ravi
````

### Usecase: Stop when the correct PIN is entered

````python
# Real-world use case: PIN verification system
# Using while True with break for input validation

# Note: In Jupyter, use Ctrl+C to interrupt if you want to stop manually
while True:
    pin = input("Enter PIN: ")
    if pin == "1234":
        print("✅ Access Granted")
        break  # Exit the infinite loop when correct PIN is entered
    print("❌ Incorrect PIN, Try Again.")

# This pattern is common in:
# - Login systems
# - Menu-driven programs
# - Input validation loops
````

### Nested Loops

#### Example: Multiplication Table

````python
# Nested Loops Example 1: Multiplication Table
# Outer loop: Controls rows (multiplicand)
# Inner loop: Controls columns (multiplier)

for i in range(1, 4):  # Rows: 1, 2, 3
    for j in range(1, 6):  # Columns: 1, 2, 3, 4, 5
        print(i * j, end="\t")  # \t creates tab spacing
    print()  # New line after each row

# Output creates a 3x5 table:
# 1  2  3  4  5
# 2  4  6  8  10
# 3  6  9  12 15
````

### Example: Pattern Matching

````python
# Nested Loops Example 2: Pattern Printing
# Creates a triangular star pattern

for i in range(1, 6):  # 5 rows
    for j in range(i):  # Print i stars in row i
        print("*", end="")  # No new line between stars
    print()  # New line after each row

# Output:
# *
# **
# ***
# ****
# *****

# Try: Modify to create inverted pattern or pyramid!
````

### `range()` Function - Deep Dive

range() is commonly used in `for` loops

````python
# Understanding range() function - generates sequence of numbers

# 1. Single argument: range(stop) - starts from 0
print("range(5):", list(range(5)))  # [0, 1, 2, 3, 4]

# 2. Two arguments: range(start, stop)
print("range(2, 6):", list(range(2, 6)))  # [2, 3, 4, 5]

# 3. Three arguments: range(start, stop, step)
print("range(1, 10, 2):", list(range(1, 10, 2)))  # [1, 3, 5, 7, 9]

# Pro tip: step can be negative for countdown!
print("Countdown:", list(range(10, 0, -1)))  # [10, 9, 8, ..., 1]
````

### `zip()` - Loop Through Multiple Lists Togeather

````python
# zip() - Iterate through multiple lists simultaneously
# Pairs up elements from multiple sequences

names = ["Aarav", "Divya", "Meera"]
marks = [85, 90, 78]

# Instead of using indices: names[i] and marks[i]
for name, mark in zip(names, marks):
    print(name, "scored", mark)

# Output:
# Aarav scored 85
# Divya scored 90
# Meera scored 78

# Note: zip stops at the shortest list
````

### `enumerate()` - Loop with Index

````python
students = ["Aarav", "Divya", "Meera"]
````

````python
# enumerate() - Loop with automatic index tracking
# Returns both index and value in each iteration

students = ["Aarav", "Divya", "Meera"]

# Instead of: for i in range(len(students))
for index, name in enumerate(students):
    print(f"{index+1}. {name}")  # Using f-string for formatting

# Output:
# 1. Aarav
# 2. Divya
# 3. Meera

# You can also start from different index:
# enumerate(students, start=1)
````

````python
# Practical use case: Find failing students with their position

marks = [78, 82, 51, 66]

for i, score in enumerate(marks):
    if score < 70:
        print(f"⚠️ Student {i + 1} scored below passing marks: {score}")

# enumerate is useful when you need both:
# 1. The position/index (i)
# 2. The value (score)
````

### Example: Update a value at Specific Index

````python
# Advanced: Update list values at specific index using enumerate()

fruits = ["apple", "banana", "grape"]

for i, fruit in enumerate(fruits):
    if fruit == "banana":
        fruits[i] = "mango"  # Replace banana with mango
        print(f"Replaced {fruit} at index {i} with mango")
````

````python
# Verify the list was modified
print("Updated fruits list:", fruits)

# Note: This demonstrates how enumerate gives you
# both read access (fruit) and write access (fruits[i])
````

# 🎊 Session Summary: Advanced Loop Control

## 📝 What We Learned

### 1️⃣ Loop Control Statements

| Statement | Purpose | When to Use |
|-----------|---------|-------------|
| `break` | Exit loop immediately | Found what you're looking for, error occurred |
| `continue` | Skip current iteration, go to next | Skip invalid data, filter items |
| `else` | Runs if loop completes normally | Detect if loop finished vs was broken |

**Example Flow:**
```python
for item in collection:
    if skip_condition:
        continue  # Skip this item
    if stop_condition:
        break     # Stop entirely
    process(item)  # Normal processing
else:
    print("All items processed")  # Only if no break occurred
```

### 2️⃣ Nested Loops

**Structure:**
```python
for outer in outer_sequence:      # Runs N times
    for inner in inner_sequence:  # Runs M times for each outer
        # This runs N × M times!
```

**Common Uses:**
- 📊 Multiplication tables
- 🎨 Pattern printing (stars, pyramids)
- 🔍 Searching 2D structures (matrices, grids)
- 🎲 Generating combinations

**Performance Note:** Be careful with deeply nested loops - time complexity grows multiplicatively!

---

### 3️⃣ Advanced Loop Functions

#### 🔢 `enumerate(sequence, start=0)`
- **Returns:** (index, value) pairs
- **Use when:** You need both position and value
- **Better than:** Manual counter or `range(len())`

```python
# ❌ Old way
for i in range(len(items)):
    print(i, items[i])

# ✅ Better way
for i, item in enumerate(items):
    print(i, item)
```

#### 🔗 `zip(*sequences)`
- **Returns:** Tuples of corresponding elements
- **Use when:** Processing parallel lists
- **Stops at:** Shortest sequence length

```python
names = ['Alice', 'Bob']
ages = [25, 30]
cities = ['NYC', 'LA']

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, from {city}")
```

#### 📏 `range(start, stop, step)`
- **Single arg:** `range(5)` → 0 to 4
- **Two args:** `range(2, 8)` → 2 to 7
- **Three args:** `range(0, 10, 2)` → 0, 2, 4, 6, 8
- **Negative step:** `range(10, 0, -1)` → countdown

## 💡 Best Practices & Tips

### When to Use Each Control Statement

| Scenario | Use |
|----------|-----|
| Stop searching once item found | `break` |
| Skip invalid/unwanted items | `continue` |
| Detect if search was successful | `for...else` |
| Need both index and value | `enumerate()` |
| Process multiple lists together | `zip()` |
| Create patterns or tables | Nested loops |

### Common Patterns

**1. Search Pattern:**
```python
for item in collection:
    if item == target:
        print("Found!")
        break
else:
    print("Not found")
```

**2. Filter Pattern:**
```python
for item in collection:
    if not valid(item):
        continue
    process(item)
```

**3. Indexed Processing:**
```python
for i, value in enumerate(data):
    if condition(value):
        data[i] = transform(value)
```
