## What is a Dictionary?

A __dictionary__ is a __collection of key-value pairs__. <br>
Each value is accessed using a __key__, not an index.

my_dict = {
    "key1":"value1",
    "key2":"value2"
}

````python
# Creating a dictionary: key-value pairs separated by colons
# Keys must be unique; values can be any data type

student = {
    "name": "Sarvesh",      # String value
    "age": 36,              # Integer value
    "course": "Python Fundamentals"  # String value
}
````

````python
student["name"]
````

### Safe access using `get()`:

````python
student.get("name", "NA")
````

### Add or update a value

````python
student['grade'] = "A"
student['age'] = 27
````

````python
# Adding new key-value pair
student['grade'] = "A"  # New key 'grade'

# Updating existing value
student['age'] = 27  # Update existing key 'age'

print("Updated student:", student)
````

````python
print(student['age'])
````

### Delete a key

````python
# Delete a key-value pair using 'del'
# Note: Raises KeyError if key doesn't exist

if 'grade' in student:  # Check before deleting to avoid errors
    del student['grade']
    print("Grade deleted successfully")
else:
    print("Grade key not found")
````

````python
print(student['grade'])
````

````python
student.pop("age")
````

````python
# .pop() removes a key and returns its value
# Safer than 'del' because you can provide a default value

removed_age = student.pop("age", "Key not found")
print(f"Removed age: {removed_age}")
print("Student after pop:", student)
````

````python
student
````

### Looping through Dictionaries

### Loop through keys 

````python
for key in student:
    print(key)
````

### Loop through keys and values

````python
for key, value in student.items():
    print(key, ":", value)
````

### Loop through values

````python
for value in student.values():
    print(value)
````

| Method            | Description                           |
| ----------------- | ------------------------------------- |
| `dict.keys()`     | Returns all keys                      |
| `dict.values()`   | Returns all values                    |
| `dict.items()`    | Returns (key, value) pairs            |
| `dict.get(key)`   | Returns value (or default if missing) |
| `dict.update({})` | Updates with another dictionary       |
| `dict.pop(key)`   | Removes key                           |
| `dict.clear()`    | Clears all items                      |


````python
employee = {
    "id": 101,
    "name": 'Anita',
    "role": "Data Analyst",
    'skills': ['Excel', 'SQL', 'Python']
}

print("Skills:", employee['skills'][1])
````

### Nested Dictionaries

````python
students = {
    "A001": {"name": "Karthik", "marks": 85},
    "A002": {"name": "Divya", "marks": 90}
}
````

````python
# Nested dictionary: dictionary containing other dictionaries
# Perfect for structured data like databases or JSON

students = {
    "A001": {"name": "Karthik", "marks": 85},
    "A002": {"name": "Divya", "marks": 90},
    "A003": {"name": "Rahul", "marks": 78}
}

# Accessing nested dictionary values
print("Student A002 marks:", students["A002"]['marks'])
print("Student A001 name:", students["A001"]['name'])
````

# Dictionary vs List vs Set vs Tuple

| Feature    | List     | Tuple    | Set   | Dictionary     |
| ---------- | -------- | -------- | ----- | -------------- |
| Syntax     | `[ ]`    | `( )`    | `{ }` | `{key: value}` |
| Ordered    | ✅        | ✅        | ❌     | ✅ (3.7+)       |
| Indexed    | By index | By index | ❌     | ✅ By key       |
| Mutable    | ✅        | ❌        | ✅     | ✅              |
| Duplicates | ✅        | ✅        | ❌     | ❌ (keys)       |

## ✅ What You've Learned Today

### 1️⃣ Dictionary Fundamentals
- **What are dictionaries**: Collections of key-value pairs
- **Syntax**: `{"key": "value"}`
- **Keys must be**: Unique and immutable (strings, numbers, tuples)
- **Values can be**: Any data type (strings, numbers, lists, dicts, etc.)

### 2️⃣ Accessing Dictionary Data
- **Square brackets**: `dict["key"]` - Raises KeyError if key missing
- **`.get()` method**: `dict.get("key", default)` - Returns default if key missing
- **Best practice**: Always use `.get()` for safe access

### 3️⃣ Dictionary Operations
- **Add new item**: `dict["new_key"] = value`
- **Update existing**: `dict["key"] = new_value`
- **Delete item**: `del dict["key"]` or `dict.pop("key")`
- **Check existence**: `"key" in dict` (returns True/False)

### 4️⃣ Looping Through Dictionaries
- **Loop keys**: `for key in dict:` (default)
- **Loop values**: `for value in dict.values():`
- **Loop both**: `for key, value in dict.items():`
- **Best practice**: Use `.items()` when you need both key and value

### 5️⃣ Essential Dictionary Methods
| Method | Purpose | Example |
|--------|---------|----------|
| `.keys()` | Get all keys | `dict.keys()` |
| `.values()` | Get all values | `dict.values()` |
| `.items()` | Get (key, value) pairs | `dict.items()` |
| `.get(key, default)` | Safe access | `dict.get("name", "Unknown")` |
| `.pop(key)` | Remove and return value | `dict.pop("age")` |
| `.update(other_dict)` | Merge dictionaries | `dict.update({"new": "value"})` |
| `.clear()` | Remove all items | `dict.clear()` |

### 6️⃣ Nested Dictionaries
- **Purpose**: Store complex, hierarchical data
- **Access pattern**: `dict["outer_key"]["inner_key"]`
- **Use cases**: User profiles, database records, JSON data

### 7️⃣ When to Use Dictionaries
- ✅ **Use dictionaries when**:
  - You need to associate keys with values (like a phonebook)
  - You want fast lookups by key
  - You're working with JSON data
  - You need to store structured data (name, age, etc.)

- ❌ **Don't use dictionaries when**:
  - You only need an ordered sequence (use list)
  - You need unique unordered items (use set)
  - You need immutable data (use tuple)

## 🎯 Key Best Practices

1. **Always use `.get()` for safe access**
   ```python
   # ✅ Good - won't crash if key missing
   value = user.get("email", "No email")
   
   # ❌ Risky - crashes with KeyError if key missing
   value = user["email"]
   ```

2. **Check key existence before deleting**
   ```python
   # ✅ Good
   if "key" in my_dict:
       del my_dict["key"]
   
   # Or use pop with default
   my_dict.pop("key", None)
   ```

3. **Use meaningful key names**
   ```python
   # ✅ Good - clear and descriptive
   user = {"first_name": "John", "last_name": "Doe", "age": 30}
   
   # ❌ Bad - unclear keys
   user = {"fn": "John", "ln": "Doe", "a": 30}
   ```

4. **Use `.items()` when looping with both key and value**
   ```python
   # ✅ Good - clean and readable
   for key, value in student.items():
       print(f"{key}: {value}")
   
   # ❌ Less efficient
   for key in student:
       print(f"{key}: {student[key]}")
   ```

5. **Remember: Keys must be immutable**
   ```python
   # ✅ Valid keys
   valid = {"name": "John", 123: "number", (1, 2): "tuple"}
   
   # ❌ Invalid - lists are mutable
   # invalid = {[1, 2]: "value"}  # TypeError!
   ```

## 💡 Common Patterns You Should Know

```python
# Pattern 1: Count occurrences
word_count = {}
for word in text.split():
    word_count[word] = word_count.get(word, 0) + 1

# Pattern 2: Grouping data
students_by_grade = {

    "B": ["Charlie", "Diana"]
}

# Pattern 3: Default values with .setdefault()
config = {"theme": "dark"}
config.setdefault("language", "English")  # Adds if not present

# Pattern 4: Dictionary comprehension
squares = {x: x**2 for x in range(5)}  # {0: 0, 1: 1, 2: 4, ...}
```
