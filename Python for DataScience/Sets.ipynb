## What is a Set?

A __set__ is an __unordered, mutable, and unindexed__ collection of __unique elements__.

1. Unordered
2. Mutable
3. Does not allow duplicates
4. Not Indexed

my_set = {1, 2, 3}

````python
# Creating an empty set - MUST use set(), not {}
# {} creates an empty dictionary, not a set!

empty_set = set()
print("Empty set:", empty_set)
print("Type:", type(empty_set))
````

````python
# Common mistake: {} creates a dictionary, NOT a set!
not_a_set = {}
print("Type of {}:", type(not_a_set))  # Will show <class 'dict'>
````

````python
# Creating sets - duplicates are automatically removed
# Sets are unordered, so output order may vary

colors = {'red', 'green', 'blue'}
print("Colors set:", colors)

# Duplicates are automatically removed
numbers = {1, 2, 3, 3, 2, 1}  # Only {1, 2, 3} will be stored
print("Numbers set (duplicates removed):", numbers)
````

### Access Elements in Sets

````python
# Sets are NOT indexed - you cannot access by position
# This will raise a TypeError

try:
    print(colors[0])  # This will fail!
except TypeError as e:
    print(f"Error: {e}")
    print("Sets don't support indexing!")
````

````python
# Looping through sets - order is not guaranteed
# Sets are unordered, so iteration order may vary

print("Colors in the set:")
for color in colors:
    print(f"  - {color}")
````

### Add Elements:

````python
# Adding elements to a set using .add()
# If element already exists, nothing happens (no error)

colors.add("yellow")
print("After adding yellow:", colors)

# Adding duplicate has no effect
colors.add("red")  # 'red' already exists
print("After adding red again:", colors)  # No change
````

````python
print(colors)
````

### Remove Elements

````python
# .remove() - removes element, raises KeyError if not found
# Use this when you're sure the element exists

if 'green' in colors:
    colors.remove("green")
    print("After removing green:", colors)
````

````python
print(colors)
````

````python
# Trying to remove non-existent element raises KeyError

try:
    colors.remove("green")  # Already removed
except KeyError:
    print("❌ KeyError: 'green' not in set")
````

````python
# .discard() - removes element, NO ERROR if not found
# Safer than .remove() - won't crash your program

colors.discard("yellow")
print("After discarding yellow:", colors)
````

````python
print(colors)
````

````python
# Discarding non-existent element - no error!
colors.discard("yellow")  # Already removed, but no error
print("No error with discard!")
````

````python
# .clear() - removes all elements from the set
colors.clear()
print("After clear:", colors)  # set()
````

````python
print(colors)
````

### Set Operations

````python
# Set Operations - Union (combines both sets)
# Union: all unique elements from both sets

a = {1, 2, 3}
b = {3, 4, 5}

# Method 1: Using | operator
print("Union (a | b):", a | b)  # {1, 2, 3, 4, 5}
````

````python
# Method 2: Using .union() method
print("Union (a.union(b)):", a.union(b))  # {1, 2, 3, 4, 5}
````

````python
# Intersection: common elements in both sets
# Using & operator
print("Intersection (a & b):", a & b)  # {3}
````

````python
# Using .intersection() method
print("Intersection (a.intersection(b)):", a.intersection(b))  # {3}
````

````python
# Difference: elements in 'a' but NOT in 'b'
print("Difference (a - b):", a - b)  # {1, 2}
````

````python
# Difference: elements in 'b' but NOT in 'a'
print("Difference (b - a):", b - a)  # {4, 5}
````

````python
# Symmetric Difference: elements in either set, but NOT in both
# (Union minus Intersection)
print("Symmetric Difference (a ^ b):", a ^ b)  # {1, 2, 4, 5}
````

| Method              | Description                         |
| ------------------- | ----------------------------------- |
| `add(item)`         | Adds item to set                    |
| `remove(item)`      | Removes item (error if not found)   |
| `discard(item)`     | Removes item (no error)             |
| `clear()`           | Empties the set                     |
| `union(set)`        | Combines two sets                   |
| `intersection(set)` | Common elements only                |
| `difference(set)`   | Elements in one but not the other   |
| `issubset(set)`     | Checks if one set is inside another |
| `issuperset(set)`   | Checks if set contains another      |


### Use a set when:

1. You want only unique items

2. Order doesn’t matter

3. You need to do set math operations

4. You want fast membership checking (`in` is faster in sets than lists)

## ✅ What You've Learned Today

### 1️⃣ Set Fundamentals
- **What are sets**: Unordered, mutable collections of **unique** elements
- **Syntax**: `{1, 2, 3}` or `set([1, 2, 3])`
- **Empty set**: Must use `set()`, NOT `{}` (that's a dictionary!)
- **Key properties**:
  - ✅ Unordered (no guaranteed order)
  - ✅ Mutable (can add/remove elements)
  - ✅ Unique (automatically removes duplicates)
  - ❌ Not indexed (cannot access by position)

### 2️⃣ Creating Sets
```python
# Method 1: Curly braces
my_set = {1, 2, 3}

# Method 2: set() constructor
my_set = set([1, 2, 3])

# Empty set (IMPORTANT!)
empty = set()  # ✅ Correct
not_set = {}   # ❌ This is a dictionary!

# Automatic duplicate removal
numbers = {1, 2, 2, 3, 3}  # Results in {1, 2, 3}
```

### 3️⃣ Adding & Removing Elements
| Operation | Method | Behavior |
|-----------|--------|----------|
| Add | `set.add(item)` | Adds single element (no error if exists) |
| Remove | `set.remove(item)` | Removes element (KeyError if not found) |
| Discard | `set.discard(item)` | Removes element (NO error if not found) |
| Pop | `set.pop()` | Removes random element |
| Clear | `set.clear()` | Removes all elements |

**Best Practice**: Use `.discard()` instead of `.remove()` to avoid errors!

### 4️⃣ Set Operations (Math Operations)
| Operation | Operator | Method | Result |
|-----------|----------|--------|--------|
| **Union** | `a \| b` | `a.union(b)` | All elements from both sets |
| **Intersection** | `a & b` | `a.intersection(b)` | Only common elements |
| **Difference** | `a - b` | `a.difference(b)` | Elements in a but not in b |
| **Symmetric Diff** | `a ^ b` | `a.symmetric_difference(b)` | Elements in either, not both |

**Visual Example**:
```python
a = {1, 2, 3}
b = {3, 4, 5}

a | b  # {1, 2, 3, 4, 5}  - Everything
a & b  # {3}              - Common only
a - b  # {1, 2}           - Only in a
b - a  # {4, 5}           - Only in b
a ^ b  # {1, 2, 4, 5}     - Not in both
```

### 5️⃣ Set Comparison Methods
| Method | Purpose | Example |
|--------|---------|----------|
| `issubset()` | Check if all elements in another set | `{1, 2}.issubset({1, 2, 3})` → True |
| `issuperset()` | Check if contains all elements of another | `{1, 2, 3}.issuperset({1, 2})` → True |
| `isdisjoint()` | Check if no common elements | `{1, 2}.isdisjoint({3, 4})` → True |

### 6️⃣ When to Use Sets

✅ **Use Sets When**:
1. **Need unique values**: Remove duplicates from a list
   ```python
   unique = list(set([1, 2, 2, 3, 3]))  # [1, 2, 3]
   ```

2. **Fast membership testing**: Check if item exists
   ```python
   if item in my_set:  # Much faster than list!
   ```

3. **Mathematical operations**: Union, intersection, difference
   ```python
   common_students = python_class & ai_class
   ```

4. **Remove duplicates**: Clean data
   ```python
   emails = set(all_emails)  # Automatic deduplication
   ```

❌ **Don't Use Sets When**:
- You need to maintain order (use list or tuple)
- You need to access by index (use list)
- You need duplicate values (use list)
- You need mutable elements (sets only accept immutable items)

## 🎯 Key Best Practices

1. **Use `.discard()` for safe removal**
   ```python
   # ✅ Good - won't crash
   my_set.discard(item)
   
   # ❌ Risky - KeyError if item not found
   my_set.remove(item)
   ```

2. **Remember: sets are unordered**
   ```python
   # Don't rely on order!
   my_set = {3, 1, 2}  # May print in any order
   ```

3. **Use sets for deduplication**
   ```python
   # ✅ Quick way to remove duplicates
   unique_items = list(set(items))
   ```

4. **Sets only accept immutable items**
   ```python
   # ✅ Valid
   valid = {1, "hello", (1, 2), True}
   
   # ❌ Invalid - lists are mutable
   # invalid = {[1, 2]}  # TypeError!
   ```

5. **Use operators for readability**
   ```python
   # ✅ Clean and readable
   result = a | b
   
   # Also valid, but longer
   result = a.union(b)
   ```

## 💡 Common Use Cases

```python
# Use Case 1: Remove duplicates from list
numbers = [1, 2, 2, 3, 3, 4]
unique = list(set(numbers))  # [1, 2, 3, 4]

# Use Case 2: Find common elements
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
common = set(list1) & set(list2)  # {3, 4}

# Use Case 3: Fast membership testing
allowed_users = set(['admin', 'user1', 'user2'])
if username in allowed_users:  # Very fast!
    print("Access granted")

# Use Case 4: Find unique items
all_tags = ['python', 'ai', 'python', 'ml', 'ai']
unique_tags = set(all_tags)  # {'python', 'ai', 'ml'}

# Use Case 5: Set algebra
python_students = {'Alice', 'Bob', 'Charlie'}
ai_students = {'Bob', 'Charlie', 'Diana'}

both = python_students & ai_students  # {'Bob', 'Charlie'}
only_python = python_students - ai_students  # {'Alice'}
all_students = python_students | ai_students  # All 4
```
