## 2. Python Operators

Operators are special symbols that perform operations on variables and values.

📝 **Student Notes:**
- Operators make Python powerful for calculations and logic
- Different operators serve different purposes
- Understanding operator precedence is crucial

### 2.1 Arithmetic Operators

Arithmetic operators perform mathematical operations.

📝 **Student Notes:**
- Use `/` for regular division (always returns float)
- Use `//` for floor division (returns integer part)
- Use `%` to get the remainder (useful for checking even/odd numbers)
- Use `**` for exponentiation (power)

| Operator | Meaning        | Example  | Result |
| -------- | -------------- | -------- | ------ |
| `+`      | Addition       | `5 + 2`  | `7`    |
| `-`      | Subtraction    | `5 - 2`  | `3`    |
| `*`      | Multiplication | `5 * 2`  | `10`   |
| `/`      | Division       | `5 / 2`  | `2.5`  |
| `//`     | Floor Division | `5 // 2` | `2`    |
| `%`      | Modulus        | `5 % 2`  | `1`    |
| `**`     | Exponentiation | `2 ** 3` | `8`    |


```python
# Addition operator
5 + 2  # Result: 7
```

```python
# Subtraction operator
5 - 3  # Result: 2
```

```python
# Division operator (always returns a float)
10 / 2  # Result: 5.0 (notice it's a float, not integer)
```

```python
# More arithmetic examples
print(f"Multiplication: 5 * 3 = {5 * 3}")
print(f"Floor Division: 17 // 5 = {17 // 5}")  # Gets the quotient
print(f"Modulus (Remainder): 17 % 5 = {17 % 5}")  # Gets the remainder
print(f"Exponentiation: 2 ** 8 = {2 ** 8}")  # 2 to the power of 8
```

### 2.2 Assignment Operators

Assignment operators are used to assign values to variables.

📝 **Student Notes:**
- `=` assigns a value to a variable
- Compound operators (`+=`, `-=`, etc.) perform an operation and assign in one step
- These operators make your code more concise
- `x += 5` is the same as `x = x + 5`

| Operator | Example  | Meaning       |
| -------- | -------- | ------------- |
| `=`      | `x = 5`  | Assign 5 to x |
| `+=`     | `x += 2` | `x = x + 2`   |
| `-=`     | `x -= 1` | `x = x - 1`   |
| `*=`     | `x *= 3` | `x = x * 3`   |
| `/=`     | `x /= 2` | `x = x / 2`   |


```python
# Basic assignment operator
x = 5  # Assign the value 5 to variable x
```

```python
# Add and assign: x = x + 2
x += 2  # x is now 7 (5 + 2)
```

```python
# Display the current value of x
print(x)  # Output: 7
```

```python
# Subtract and assign: x = x - 1
x -= 1  # x is now 6 (7 - 1)
print(x)  # Output: 6
```

```python
# Divide and assign: x = x / 2
x /= 2  # x is now 3.0 (6 / 2)
print(x)  # Output: 3.0
```

### 3. Comparison Operators

| Operator | Meaning          | Example  | Result  |
| -------- | ---------------- | -------- | ------- |
| `==`     | Equal to         | `5 == 5` | `True`  |
| `!=`     | Not equal to     | `5 != 3` | `True`  |
| `>`      | Greater than     | `5 > 3`  | `True`  |
| `<`      | Less than        | `5 < 3`  | `False` |
| `>=`     | Greater or equal | `5 >= 5` | `True`  |
| `<=`     | Less or equal    | `5 <= 4` | `False` |


```python
score = 90

print(score >= 60)
```

### 4. Logical Operators

Logical operators combine Boolean expressions.

📝 **Student Notes:**
- `and`: all conditions must be True
- `or`: at least one condition must be True
- `not`: flips True to False (and vice versa)
- Combine with comparison operators for powerful conditions

| Operator | Meaning                     | Example            |
| -------- | --------------------------- | ------------------ |
| `and`    | All conditions must be True | `x > 5 and x < 10` |
| `or`     | At least one condition True | `x < 5 or x > 15`  |
| `not`    | Reverses a condition        | `not(x > 10)`      |


```python
x = 4

if x>5 and x<10:
    print("True")
else:
    print("False")
```

```python
x = 4

if x>5 or x<10:
    print("True")
else:
    print("False")
```

```python
x = 4

not(x>10)
```

## 3. Operator Precedence

Operator precedence determines the order in which operations are performed in an expression.

📝 **Student Notes:**
- Remember PEMDAS: Parentheses, Exponents, Multiplication/Division, Addition/Subtraction
- Exponentiation (`**`) is right-to-left; most others are left-to-right
- Use parentheses `()` to make intent clear and avoid surprises

###  Python Operator Precedence (High to Low)

| Precedence Level | Operators                       | Description                          | Associativity     |
|------------------|----------------------------------|--------------------------------------|-------------------|
| 1 (Highest)      | `()`                             | Parentheses (grouping)               | -                 |
| 2                | `**`                             | Exponentiation                       | Right to Left     |
| 3                | `+x`, `-x`, `~x`                 | Unary plus, minus, bitwise NOT       | Right to Left     |
| 4                | `*`, `/`, `//`, `%`              | Multiplication, Division, Floor Div, Modulus | Left to Right |
| 5                | `+`, `-`                         | Addition and Subtraction             | Left to Right     |
| 6                | `<<`, `>>`                       | Bitwise shift operators              | Left to Right     |
| 7                | `&`                              | Bitwise AND                          | Left to Right     |
| 8                | `^`                              | Bitwise XOR                          | Left to Right     |
| 9                | `|`                              | Bitwise OR                           | Left to Right     |
| 10               | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in` | Comparison & identity/membership | Left to Right     |
| 11               | `not`                            | Logical NOT                          | Right to Left     |
| 12               | `and`                            | Logical AND                          | Left to Right     |
| 13               | `or`                             | Logical OR                           | Left to Right     |
| 14 (Lowest)      | `=`, `+=`, `-=`, `*=`, etc.      | Assignment operators                 | Right to Left     |

> 📝 Tip: Use parentheses `()` to clarify and control precedence.


```python
3 + 2 * 5
```

### Precedence is NOT Left to Right!

Operations run based on precedence rules, not just left-to-right order.

```python
print(2 + 3 * 4)
```

## 4. User Input and Type Casting

Learning to get input from users and convert data types is essential for interactive programs.

📝 **Student Notes:**
- `input()` always returns a string
- Convert to other types with `int()`, `float()`, `str()`, `bool()` as needed
- Provide clear prompts; validate user input where possible

```python
name = input("Enter your name")

print(name)
```

### Type Casting

```python
age = int(input("Enter your age"))

print(age)
```

```python
type(age)
```

### Convert an integer to string

```python
age = 25

message = "I am " + str(age) + " years old."

print(message)
```

### Convert String to Boolean

```python
user_input = input("type 'yes' to subscribe: ")

is_subscribed = bool(user_input.lower() == 'yes')

print("Subscription active:", is_subscribed)
```

### Multiple Inputs in One Line

```python
x, y = input("Enter two numbers seperated by space: ").split()
```

```python
print(x, y)
```

### Password Masking(Optional with getpass)

```python
import getpass

password = getpass.getpass("Enter your password: ")

print("Password received")
```

```python
import getpass

username = input("Enter username: ")
password = getpass.getpass("Enter your password: ")

print("Login attempt received for user:", username)
```

## 5. Print Formatting

Modern Python offers powerful string formatting techniques to create clean, readable output.

📝 **Student Notes:**
- Prefer **f-strings** for clarity and speed (prefix with `f"..."`)
- Use `{}` placeholders to inject variables or expressions
- `.format()` works but is more verbose—know it for legacy code
- Double-check that you didn’t forget the leading `f` (common beginner bug)

```python
name = "sarvesh"

age = 34

print(f"Hello {name}, you are {age} years old")
```

```python
print(f"Hello name, you are age years old")
```

```python
a = 5
b = 10

print(f"The sum of {a} and {b} is {a + b}")
```

### Using `format()` Method

```python
name = "sarvesh"

age = 34

print("Hello {0}, you are {1} years old".format(name, age))
```

- `f-strings` - Modern, clean, readable
- `.format()` - Compatible with older versions
