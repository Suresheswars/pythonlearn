# Part 1: NumPy - Numerical Python ⚡

## 📦 What is NumPy?

NumPy (Numerical Python) is the foundation of scientific computing in Python. It provides:
- **N-dimensional arrays** - Store large amounts of data efficiently
- **Vectorized operations** - Perform calculations 100x faster than Python loops
- **Mathematical functions** - Built-in support for linear algebra, statistics, etc.

### 🎯 Why NumPy?

| Feature | Python List | NumPy Array |
|---------|-------------|-------------|
| Speed | Slow (loop-based) | **100x Faster** |
| Memory | High | **Low (50% less)** |
| Operations | One-by-one | **Vectorized** |
| Data Types | Mixed | Homogeneous (same type) |

---

## 🔢 Creating NumPy Arrays

Arrays come in different dimensions:
- **1D Array** - Like a list: `[1, 2, 3, 4]`
- **2D Array** - Like a table: `[[1, 2], [3, 4]]`
- **3D+ Array** - For images, videos, complex data


````python
# Import NumPy - the standard alias is 'np'
import numpy as np

# 1D Array: Single row of data (like a Python list, but faster)
arr1 = np.array([1, 2, 3, 4])
print("1D Array:", arr1)

# 2D Array: Table-like structure with rows and columns
# Think of it as a matrix or spreadsheet
arr2 = np.array([[1, 2], [3, 4]])
print("\n2D Array:")
print(arr2)

# Note: All elements must be the same data type (homogeneous)
# NumPy automatically converts to the most appropriate type
````

````python
print(arr1.shape)

print(arr2.shape)
````

````python
print(arr1[0])
````

````python
print(arr2[0][0])
````

## ⚡ NumPy Math Operations - The Power of Vectorization

### Why Vectorization Matters

Instead of looping through each element (slow), NumPy applies operations to **entire arrays at once** (fast)!

**Example:**
```python
# Python way (SLOW) ❌
result = []
for x in [1, 2, 3]:
    result.append(x * 2)

# NumPy way (FAST) ✅
result = np.array([1, 2, 3]) * 2
```

### Common Operations


````python
# Create an array of numbers
arr = np.array([10, 20, 30])

# SCALAR OPERATIONS: Apply operation to every element

# Addition: Add 5 to each element
print("Original:", arr)
print("Add 5:", arr + 5)  # [15, 25, 35]

# Multiplication: Multiply each element by 5
print("Multiply by 5:", arr * 5)  # [50, 100, 150]

# Division: Divide each element by 10
print("Divide by 10:", arr / 10)  # [1.0, 2.0, 3.0]

# Note: Original array 'arr' is NOT modified (NumPy creates new arrays)
````

````python
np.sum(arr)
````

````python
np.mean(arr)
````

````python
a = np.array([10, 20, 30])

b = np.array([1, 2, 3])
````

````python
c = a + b
````

## 🔄 Reshape Arrays - Change Dimensions

### Why Reshape?

Machine learning algorithms often require specific shapes:
- **Images**: 1D array → 2D/3D (height × width × channels)
- **Models**: Must match input/output dimensions
- **Data preprocessing**: Transform data for analysis

### Rules for Reshaping

✅ **Total elements must remain the same**
- 6 elements: (6,) → (2, 3) ✓
- 6 elements: (6,) → (3, 2) ✓
- 6 elements: (6,) → (2, 4) ✗ (needs 8 elements)

Formula: `rows × columns = total elements`


````python
import numpy as np

# Original 1D array with 6 elements
a = np.array([1, 2, 3, 4, 5, 6])

# Reshape to 2 rows × 3 columns (2×3 = 6 elements ✓)
b = a.reshape(2, 3)

# You can also reshape to:
c = a.reshape(3, 2)  # 3 rows × 2 columns
d = a.reshape(6, 1)  # Column vector (6 rows × 1 column)
````

````python
print("Original:", a)

print("Reshaped (2x3):\n", b)
````

## 📡 Broadcasting - NumPy's Superpower

### What is Broadcasting?

Broadcasting allows NumPy to perform operations on arrays of **different shapes** without copying data!

**Example:**
```python
# Array: (2×3)     [[1, 2, 3],
#                   [4, 5, 6]]
#
# Vector: (3,)     [10, 20, 30]
#
# Result: (2×3)    [[11, 22, 33],
#                   [14, 25, 36]]
```

### Broadcasting Rules

1. If arrays have different dimensions, **pad smaller array's shape with 1s** on the left
2. Arrays are compatible if dimensions are **equal OR one of them is 1**
3. Arrays with dimension 1 are **stretched** to match the other array

### Common Use Cases

- Add a constant to each row
- Normalize data (subtract mean, divide by std)
- Image processing (apply filters)


````python
# 2D array (2 rows × 3 columns)
a = np.array([[1, 2, 3], [4, 5, 6]])

# 1D array (3 elements) - will be broadcast to match 'a'
b = np.array([10, 20, 30])

# NumPy automatically broadcasts 'b' to each row of 'a'
# Row 1: [1, 2, 3] + [10, 20, 30] = [11, 22, 33]
# Row 2: [4, 5, 6] + [10, 20, 30] = [14, 25, 36]
print("Result of broadcasting:")
print(a + b)

# Without broadcasting, you'd need nested loops! 🐌
````

## 📊 Pandas Series - One-Dimensional Labeled Array

### What is a Series?

A Series is like a **single column from Excel** with:
- **Values** - The actual data
- **Index** - Row labels (can be custom or auto-generated)

### When to Use Series?

✅ Single variable (e.g., student marks, stock prices)  
✅ Time series data (dates as index)  
✅ Extracting a column from a DataFrame  


````python
# Create a Series with custom index (student names)
s1 = pd.Series(
    [85, 90, 78, 92],  # Values (marks)
    index=["Aarav", "Meera", "Divya", "Ravi"]  # Custom labels
)

print(s1)

# Access by label: s1['Aarav'] → 85
# Access by position: s1[0] → 85
````

## 📊 Grouping Data with `groupby()` - Split-Apply-Combine

### What is GroupBy?

`groupby()` answers questions like:
- What's the **average salary per department**?
- How many **students in each class**?
- What's the **total sales by region**?

### The Split-Apply-Combine Pattern

```
1. SPLIT     → Group data by category (e.g., by 'Grade')
2. APPLY     → Calculate something for each group (mean, count, sum)
3. COMBINE   → Merge results into a summary table
```

### Common Aggregations

| Function | Purpose | Example |
|----------|---------|----------|
| `.count()` | Count non-null values | Students per class |
| `.mean()` | Average | Average marks per grade |
| `.sum()` | Total | Total sales per region |
| `.min()` / `.max()` | Min/Max values | Youngest/oldest in group |
| `.agg()` | Multiple aggregations | Mean + Max + Min together |

### Real-World Use Cases

📈 **Business Analytics**
```python
sales_df.groupby('Region')['Revenue'].sum()
```

🎓 **Education**
```python
students_df.groupby('Class')['Marks'].mean()
```

💰 **Finance**
```python
transactions_df.groupby('Category')['Amount'].agg(['sum', 'mean', 'count'])
```


````python
df[['Grade', 'Marks']].groupby("Grade").count()
````

````python
summary = df.groupby("Class").agg({
    "Marks": ["mean", "max", "min"],
    "Name":"count"
})

print(summary)
````
