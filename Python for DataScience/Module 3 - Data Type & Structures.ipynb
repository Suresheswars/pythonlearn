## 📋 Part 1: Lists - Mutable Ordered Collections

Lists are Python's most versatile data structure for storing sequences of items.


### 🔍 What is a List?

A **list** is an ordered collection of items that can store multiple values in a single variable.

**Key Properties:**
- ✅ **Ordered**: Items maintain their position
- ✅ **Mutable**: Can be changed after creation
- ✅ **Allows duplicates**: Same value can appear multiple times
- ✅ **Mixed types**: Can contain different data types


### 📝 List Syntax

```python
my_list = [item1, item2, item3]
```

Lists use **square brackets** `[]`


````python
# Creating a List: Ordered collection of items
# Lists use square brackets []

fruits = ["apple", "banana", "cherry"]
#        index 0   index 1    index 2

# Lists can contain:
# - Strings, numbers, booleans, or mixed types
# - Duplicate values are allowed
# - Items maintain their order

print("My fruit list:", fruits)
print("Number of items:", len(fruits))  # Get list length
````

````python
# Display the list
print(fruits)
print("Type:", type(fruits))  # <class 'list'>
````

### 🎯 Accessing List Items

Access individual elements using their index (position).


````python
# Positive Indexing: Access from the start (0-based)
# First element is at index 0

print("First fruit:", fruits[0])   # apple
print("Type:", type(fruits[0]))    # <class 'str'>
````

````python
# Access second element
print("Second fruit:", fruits[1])  # banana
````

````python
# Negative Indexing: Access from the end
# -1 is the last element, -2 is second-to-last

print("Last fruit:", fruits[-1])   # cherry
````

````python
# Second-to-last element
print("Second-to-last:", fruits[-2])  # banana

# Visual representation:
# fruits = ["apple", "banana", "cherry"]
#           0        1         2        ← Positive indices
#          -3       -2        -1        ← Negative indices
````

---

### ✂️ Slicing Lists

Extract a portion of a list using slice notation.


### 📐 Slicing Syntax

```python
list[start:stop:step]
```

- **start**: Index to begin (inclusive)
- **stop**: Index to end (exclusive)
- **step**: Skip interval (default: 1)


````python
# Slicing: Extract a portion of the list
# Syntax: list[start:stop:step]

colors = ['red', 'green', 'blue', 'yellow', 'purple']
#         0       1        2       3         4

# Get elements from index 1 to 3 (stop is exclusive)
print("Colors [1:4]:", colors[1:4])
# Output: ['green', 'blue', 'yellow']

# Remember: stop index is NOT included in the result
````

````python
# Slicing: First N elements
# [:3] means from beginning to index 3 (exclusive)

print("First 3 colors:", colors[:3])
# Output: ['red', 'green', 'blue']
````

````python
# Slicing: From index to end
# [2:] means from index 2 to the end

print("From index 2 onwards:", colors[2:])
# Output: ['blue', 'yellow', 'purple']
````

````python
# Practical Example: Split data into two parts

data = [10, 20, 30, 40, 50]

first_half = data[0:3]   # First 3 items: [10, 20, 30]
second_half = data[3:5]  # Last 2 items: [40, 50]

# This is useful for dividing datasets!
````

````python
# Display both halves
print("First half:", first_half)
print("Second half:", second_half)

# Notice: Together they form the complete original list
````

````python
# Understanding how slicing works with range()
# This helps visualize the stop-exclusive behavior

for i in range(2, 5):
    print(i)

# Output: 2, 3, 4 (stops before 5)
# Same logic applies to slicing: [2:5] gives indices 2, 3, 4
````

````python
# Slicing with STEP: Skip elements
# [start:stop:step]

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# [::2] means: all elements, skip every 2nd
print("Every 2nd element:", numbers[::2])
# Output: [1, 3, 5, 7]
````

````python
# Start from index 1, then every 2nd element
print("Every 2nd from index 1:", numbers[1::2])
# Output: [2, 4, 6, 8]
````

````python
# Reverse a list using negative step!
# [::-1] means: all elements, step backwards

print("Reversed:", numbers[::-1])
# Output: [8, 7, 6, 5, 4, 3, 2, 1]

# This is a Pythonic way to reverse - no need for a function!
````

````python
# Negative indices in slicing
# Extract last 3 elements (excluding the very last)

print("Slice [-4:-1]:", numbers[-4:-1])
# Output: [5, 6, 7] (elements at positions -4, -3, -2)
````

````python
# Modifying lists through slice assignment
# You can replace multiple elements at once!

nums = [0, 1, 2, 3, 4, 5, 6]
print("Original:", nums)

# Replace elements at indices 2, 3, 4 with new values
nums[2:5] = [20, 30, 40]

print("After modification:", nums)
# Output: [0, 1, 20, 30, 40, 5, 6]
````

````python
# Delete entire list from memory
# WARNING: After this, the variable no longer exists!

del nums
print("List deleted successfully")
````

````python
# Trying to access a deleted variable causes NameError
# Uncomment to see the error:

# nums  # This will raise: NameError: name 'nums' is not defined

print("Variable 'nums' no longer exists after deletion")
````

## 🔒 Part 2: Tuples - Immutable Ordered Collections

Tuples are like lists, but they cannot be changed after creation.


### 🔍 What is a Tuple?

A **tuple** is an **ordered, immutable** collection of elements.

**Key Properties:**
- ✅ **Ordered**: Items maintain their position
- 🔒 **Immutable**: Cannot be modified after creation
- ✅ **Allows duplicates**: Same value can appear multiple times
- ⚡ **Faster**: Slightly better performance than lists

**When to use:**
- Fixed data (coordinates, RGB values, database records)
- Data that shouldn't be accidentally modified
- Dictionary keys (tuples can be keys, lists cannot!)


### 📝 Tuple Syntax

```python
my_tuple = (item1, item2, item3)
```

Tuples use **parentheses** `()`


````python
# Creating a Tuple: Immutable ordered collection
# Tuples use parentheses ()

colors = ("red", "green", "blue")
#         0       1        2

print("Color tuple:", colors)
print("Type:", type(colors))  # <class 'tuple'>
````

````python
# Accessing tuple elements (same as lists)
print("First color:", colors[0])  # red
````

````python
# Attempting to modify a tuple causes TypeError
# Tuples are IMMUTABLE - cannot be changed!

# Uncomment to see the error:
# colors[0] = "violet"  # TypeError: 'tuple' object does not support item assignment

print("⚠️ Tuples cannot be modified after creation!")
````

````python
# Tuples CAN contain duplicate values
colors = ("red", "green", "blue", "red")

print("Tuple with duplicates:", colors)
# 'red' appears twice - this is allowed
````

````python
print(colors)
````

````python
# Tuples can store different data types
# Perfect for representing a record with mixed fields

person = ("Sarvesh", 34, True)
#         name      age   is_active

print("Person record:", person)
# Output: ('Sarvesh', 34, True)
````

````python
print(person)
````

````python
# Single-element Tuple: MUST include trailing comma!
# This is a common gotcha for beginners

single_element = ("apple",)  # ← Notice the comma!
#                         ^
#                    Without comma, it's just a string

print("Single-element tuple:", single_element)
print("Type:", type(single_element))  # <class 'tuple'>
````

````python
print(single_element)
````

````python
# WITHOUT trailing comma: Not a tuple!
single_element_wo_comma = ("cherry")  # No comma = just a string

print("Without comma:", single_element_wo_comma)
print("Type:", type(single_element_wo_comma))  # <class 'str'> ❌

# Key takeaway: Always use trailing comma for single-element tuples!
````

````python
print(single_element_wo_comma)
print("This is a string, not a tuple!")
````

````python
# Slicing works with tuples too!
print("First 2 colors:", colors[0:2])
# Output: ('red', 'green')
````

### 🛠️ Tuple Methods

Tuples have only 2 methods (because they're immutable).


#### 1. `count()` - Count occurrences


````python
# Tuple Method 1: count()
# Returns how many times a value appears

nums = (1, 2, 3, 2, 2)

count_of_2 = nums.count(2)
print(f"The number 2 appears {count_of_2} times")
# Output: The number 2 appears 3 times
````

#### 2. `index()` - Find position


````python
# Tuple Method 2: index()
# Returns the FIRST index where value is found

first_position = nums.index(2)
print(f"First occurrence of 2 is at index: {first_position}")
# Output: First occurrence of 2 is at index: 1
````

### 🔄 Looping Through Tuples


````python
# Create a tuple for iteration
fruits = ("apple", "banana", "cherry")
print("Fruits tuple:", fruits)
````

````python
print(fruits)
````

````python
# Looping through a tuple
# Same syntax as looping through a list

print("\nIterating through fruits:")
for fruit in fruits:
    print(f"  - {fruit}")

# Even though tuples are immutable, you can still iterate!
````

### 📦 Tuple Packing and Unpacking

A powerful feature for working with multiple values efficiently.


````python
# Tuple Packing: Assign multiple values to a tuple
# Values are "packed" into a single tuple

person = ("Sarvesh", 34, "India")
#         name      age  country

print("Packed tuple:", person)
````

````python
# Tuple Unpacking: Extract values into separate variables
# Number of variables must match number of tuple elements!

name, age, country = person
#     ^    ^    ^
#     |    |    └─ Gets "India"
#     |    └────── Gets 34
#     └─────────── Gets "Sarvesh"

print(f"Name: {name}, Age: {age}, Country: {country}")
````

````python
# All three variables now contain individual values
print("Name:", name)
print("Age:", age)
print("Country:", country)
````

````python
# Access individual unpacked variable
print(f"Person is from: {country}")
````

## 📝 What We Learned

### 1️⃣ Lists: Mutable Ordered Collections

**Syntax:** `my_list = [item1, item2, item3]`

**Key Features:**
- ✅ **Ordered**: Elements maintain their position
- ✅ **Mutable**: Can add, remove, or modify items
- ✅ **Duplicates allowed**: Same value can appear multiple times
- ✅ **Mixed types**: Can contain strings, numbers, booleans, etc.

**Common Operations:**
```python
fruits = ['apple', 'banana', 'cherry']

# Indexing
fruits[0]       # 'apple' (first item)
fruits[-1]      # 'cherry' (last item)

# Slicing
fruits[0:2]     # ['apple', 'banana']
fruits[::-1]    # Reverse the list

# Length
len(fruits)     # 3
```

---

### 2️⃣ Slicing: Extracting Sub-sequences

**Syntax:** `sequence[start:stop:step]`

| Pattern | Example | Result | Use Case |
|---------|---------|--------|----------|
| `[start:stop]` | `nums[1:4]` | Elements 1-3 | Get range |
| `[:n]` | `nums[:3]` | First n items | Get beginning |
| `[n:]` | `nums[2:]` | From n to end | Get ending |
| `[::step]` | `nums[::2]` | Every nth item | Skip elements |
| `[::-1]` | `nums[::-1]` | Reversed | Reverse sequence |
| `[-n:]` | `nums[-3:]` | Last n items | Get tail |

**Remember:**
- 🔢 `start` is **inclusive** (included in result)
- 🚫 `stop` is **exclusive** (not included in result)
- ⚡ Negative step reverses direction

---

### 3️⃣ Tuples: Immutable Ordered Collections

**Syntax:** `my_tuple = (item1, item2, item3)`

**Key Features:**
- ✅ **Ordered**: Elements maintain their position
- 🔒 **Immutable**: **Cannot** be modified after creation
- ✅ **Duplicates allowed**: Same value can appear multiple times
- ⚡ **Faster**: Better performance than lists for reading

**Tuple Methods (Only 2!):**
```python
nums = (1, 2, 3, 2, 2)

nums.count(2)     # 3 (appears 3 times)
nums.index(2)     # 1 (first occurrence at index 1)
```

**Special Cases:**
```python
# Single-element tuple (requires trailing comma!)
single = ("apple",)    # ✅ Tuple
not_tuple = ("apple")  # ❌ String, not tuple!

# Empty tuple
empty = ()
```

---

### 4️⃣ Tuple Packing & Unpacking

**Packing: Combine values into a tuple**
```python
person = ("Alice", 25, "USA")
```

**Unpacking: Extract values into variables**
```python
name, age, country = person
# name = "Alice", age = 25, country = "USA"
```

**Advanced Unpacking with `*`:**
```python
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
# first = 1, middle = [2, 3, 4], last = 5
```

**Common Use Cases:**
- Swapping variables: `a, b = b, a`
- Returning multiple values from functions
- Iterating through paired data

## 🎯 Lists vs Tuples: When to Use Each

### Use **Lists** When:
- ✅ Data will change (add/remove items)
- ✅ Need many modification operations
- ✅ Building dynamic collections

**Examples:**
- Shopping cart items
- To-do list tasks
- User input collection
- Data processing pipelines

### Use **Tuples** When:
- ✅ Data should NOT change (immutable)
- ✅ Need slightly better performance
- ✅ Using as dictionary keys
- ✅ Protecting data integrity

**Examples:**
- Coordinates: `(x, y, z)`
- RGB colors: `(255, 128, 0)`
- Database records
- Configuration settings
- Function return values

---

## 💡 Best Practices & Tips

### Indexing & Slicing
- 🎯 Use **negative indices** for accessing from the end: `my_list[-1]`
- ✂️ Remember: `[start:stop]` **excludes** the stop index
- 🔄 Use `[::-1]` to reverse (cleaner than `reversed()`)
- 📊 Slicing creates a **new** list/tuple (doesn't modify original)

### Lists
- 📝 Check length before accessing: `if len(my_list) > 0:`
- 🔍 Use `in` for membership testing: `if "apple" in fruits:`
- ⚠️ Avoid modifying list while iterating over it

### Tuples
- ⚠️ **Always** use trailing comma for single-element tuples: `(item,)`
- 🔒 Use tuples for data that shouldn't change
- 📦 Tuple unpacking makes code cleaner and more readable
- 💾 Tuples can be dictionary keys; lists cannot
