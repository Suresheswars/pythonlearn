## 2. Markdown in Jupyter Notebooks

Markdown is a lightweight markup language that allows you to format text in Jupyter notebooks. It's perfect for adding documentation, notes, and explanations to your code.

📝 **Student Notes:**
- Markdown cells are for text, not code
- Use markdown to document your work and add explanations
- Switch cell type using the dropdown menu or keyboard shortcuts

### 2.1 Headers in Markdown

Use the `#` symbol to create headers. More `#` symbols = smaller header.

**Examples:**

# Heading 1 (Main Title)
## Heading 2 (Subtopic)
### Heading 3 (Section)
#### Heading 4 (Subsection)

### 2.2 Text Formatting

You can make text **bold**, *italic*, or ~~strikethrough~~:

📝 **Student Notes:**
- Use `**text**` or `__text__` for **bold**
- Use `*text*` or `_text_` for *italic*
- Use `~~text~~` for ~~strikethrough~~

This text is normal

This text is **bold**

This text is *italic*

This text is ***bold and italic***

~~This text is strikethrough~~

### 2.3 Lists

**Bullet Lists (Unordered):**
Use `-`, `*`, or `+` for bullet points

**Example:**

- Python
- Data Science
- Machine Learning
    - Supervised Learning
    - Unsupervised Learning
- Deep Learning

**Numbered Lists (Ordered):**
Use numbers followed by a period (1., 2., 3., etc.)

**Example:**

1. Install Python
2. Install Jupyter Notebook
3. Start learning Python
4. Build amazing projects!

### 2.4 Code in Markdown

You can display code in markdown without executing it using backticks.

**Inline code:** Use single backticks around the code

Example: Use the `print()` function to display output in Python.

**Code block:** Use triple backticks for multi-line code

```python
# This is a code block
name = "Student"
print(f"Hello, {name}!")
```

### 2.5 Block Quotes

Use `>` to create block quotes (great for highlighting important notes!)

> 💡 **Important:** This is a single-level quote

>> 📌 **Note:** This is a double-level quote (nested)

>>> ⚡ **Remember:** This is a triple-level quote (deeply nested)

### 2.6 Links and Images

**Links:** `[Display Text](URL)`

**Images:** `![Alt Text](Image URL)`

**Example Links:**

[Learn Python - Official Documentation](https://www.python.org)

[Google](https://www.google.com)

**Example Image:**

![Python Logo](https://www.python.org/static/community_logos/python-logo.png)

### 2.7 Emojis

You can add emojis directly in markdown to make your notebooks more engaging! 🎉

**Examples:**

## 4. Python Variables

Variables are containers that store data values. In Python, you don't need to declare the variable type.

📝 **Student Notes:**
- Variable names should be descriptive
- Use lowercase with underscores for multi-word variables (snake_case)
- Variables are case-sensitive: `name` and `Name` are different!
- Cannot start with a number or contain special characters (except underscore)

````python
# Creating variables - Python automatically determines the type
name = "Sarvesh"  # String (text)
age = 25  # Integer (whole number)
height = 5.9  # Float (decimal number)
is_student = True  # Boolean (True/False)

# Printing variables
print(name)
print(age)
print(height)
print(is_student)
````

````python
# You can change variable values
score = 85
print("Initial score:", score)

score = 95  # Reassigning a new value
print("Updated score:", score)
````

````python
# Multiple assignments in one line
x, y, z = 10, 20, 30
print(f"x = {x}, y = {y}, z = {z}")

# Same value to multiple variables
a = b = c = 100
print(f"a = {a}, b = {b}, c = {c}")
````

### 4.1 Data Types

Python has several built-in data types:
- **String (`str`)**: Text data - `"Hello"`, `'Python'`
- **Integer (`int`)**: Whole numbers - `42`, `100`, `-5`
- **Float (`float`)**: Decimal numbers - `3.14`, `2.5`, `-0.5`
- **Boolean (`bool`)**: True or False values

You can check the type using the `type()` function.

````python
# Checking data types using type() function
course_name = "Python for Data Science"
num_students = 100
average_rating = 4.8
is_online = True

print("Type of course_name:", type(course_name))
print("Type of num_students:", type(num_students))
print("Type of average_rating:", type(average_rating))
print("Type of is_online:", type(is_online))
````

````python
# String concatenation (combining strings)
first_name = "Learn"
last_name = "WithSarvesh"
full_name = first_name + last_name  # Using + operator
print(full_name)

# Better formatting with space
full_name_spaced = first_name + " " + last_name
print(full_name_spaced)
````

````python
# F-strings - Modern way to format strings (Python 3.6+)
name = "Student"
age = 22
city = "Mumbai"

# Using f-string for cleaner output
message = f"My name is {name}, I am {age} years old, and I live in {city}."
print(message)

# You can also do calculations inside f-strings
price = 100
quantity = 3
print(f"Total cost: {price * quantity} rupees")
````

## 5. Basic Arithmetic Operations

Python can perform mathematical operations just like a calculator!

📝 **Student Notes:**
- `+` Addition
- `-` Subtraction
- `*` Multiplication
- `/` Division (always returns a float)
- `//` Floor Division (returns integer part)
- `%` Modulus (returns remainder)
- `**` Exponentiation (power)

````python
# Basic arithmetic operations
a = 15
b = 4

print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Floor Division: {a} // {b} = {a // b}")
print(f"Modulus: {a} % {b} = {a % b}")
print(f"Exponentiation: {a} ** 2 = {a ** 2}")
````

````python
# Practical example: Calculate area of a rectangle
length = 10
width = 5
area = length * width

print(f"Rectangle dimensions: {length} x {width}")
print(f"Area of rectangle: {area} square units")

# Calculate circle area
radius = 7
pi = 3.14159
circle_area = pi * (radius ** 2)

print(f"\nCircle radius: {radius}")
print(f"Area of circle: {circle_area:.2f} square units")  # :.2f rounds to 2 decimal places
````
