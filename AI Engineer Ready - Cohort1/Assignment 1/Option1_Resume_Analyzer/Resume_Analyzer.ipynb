```python
# TODO: Implement identify_missing_skills(current_skills, target_skills)
# Hints:
# - Accept a target_skills list representing the role (e.g., ['python', 'sql', 'ml'])
# - Return which target skills are missing from current_skills

def identify_missing_skills(current_skills: list, target_skills: list) -> list:
    current = set(s.lower().strip() for s in current_skills)
    target = [t.lower().strip() for t in target_skills]
    return [t for t in target if t not in current]

print(identify_missing_skills(['python', 'sql'], ['python', 'sql', 'machine learning']))
```
