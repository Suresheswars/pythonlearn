## Part 1: Memory Basics

Memory is how a chatbot remembers previous messages. Without memory, every question feels brand new.

Think of it like this:
- **Buffer memory** = keeps everything
- **Summary memory** = keeps a short summary
- **Window memory** = keeps only the latest messages
