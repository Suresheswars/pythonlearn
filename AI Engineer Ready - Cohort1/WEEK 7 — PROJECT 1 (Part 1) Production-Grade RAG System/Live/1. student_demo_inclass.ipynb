```python
from src.backend.api import routes,health # main APIs, /health
from src.backend.logger import GLOBAL_LOGGER as log
```

```python
app = FastAPI()
```

```python
app.add_middleware(
    CORSMiddleware,
    allow_origins= ["*"], # who can access - ["https://localhost:3000"]
    allow_credentials=True, # Authentication tokens, cookies
    allow_methods=["*"], # Methods? - ["GET", "POST", "PUT", "DELETE"]
    allow_headers=["*"], # Authorization/ Content-type
)
```

```python
@app.middleware("http")
async def log_requests(request: Request, call_next):
    start_time = time.time()
    
    response = await call_next(request) 
    
    process_time = time.time() - start_time # how long each api call takes
    
    log.info(
        "Request processed",
        method=request.method,
        path=request.url.path,
        status_code=response.status_code,
        duration=f"{process_time:.4f}s"
    )
    
    return response
```

```python
app.include_router(routes.router)
app.include_router(health.router)

```
