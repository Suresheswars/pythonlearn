### Git Essentials: `diff` and `restore` (Safe Undo)

What this does:
- `git diff` shows exactly what changed before you commit.
- `git restore <file>` discards uncommitted changes in that file.

Try this quickly:
```bash
# 1) Edit notes.txt manually and add one extra line
git status
git diff
git restore notes.txt
git status
```

Use this when you made a mistake and want to go back before committing.

### Commit Message Basics

What this does:
- A good commit message explains what changed and why.

Examples:
- Good: `Add Week 1 notes with Python and Git summary`
- Weak: `update`

Rule of thumb: start with a verb like Add, Fix, Update, Remove.

### Collaboration Preview: Branch + Pull Request (Basic)

What this does:
- A branch lets you work on changes safely without touching main.
- A pull request (PR) is how you request review before merging.

Try this flow:
```bash
git switch -c feature/add-note-line
# edit notes.txt and add one line
git add notes.txt
git commit -m "Add one line in notes for practice"
git branch
```

After your remote is set:
```bash
git push -u origin feature/add-note-line
```

Then on GitHub: open Compare and pull request, add a title, and create the PR.

## Push to GitHub

1. Create a new empty repository on GitHub (for example: `week1-python-git`).
2. Connect local repo to remote:

```bash
git remote add origin https://github.com/<your-username>/week1-python-git.git
git branch -M main
git push -u origin main
```

3. Refresh GitHub and confirm your commit is visible.

## Quick Reference: When to Use Which Git Command

Save this for later!

| Command | When to use | Example |
|---------|-------------|----------|
| `git init` | Start a new project | First time in a folder |
| `git status` | Check what changed | Before you commit |
| `git diff` | See exact changes | Review edits line-by-line |
| `git add <file>` | Stage files for commit | Before committing |
| `git commit -m "msg"` | Save changes | After staging |
| `git restore <file>` | Undo uncommitted changes | Mistake before commit |
| `git log --oneline` | View commit history | See what was done |
| `git switch -c branch` | Create new branch | Start separate work |
| `git push -u origin branch` | Send branch to GitHub | Before creating PR |
| `git pull` | Get latest from remote | Sync with team |

**Golden Rule:** status → diff → add → commit → push
